## 🗃️ Block 2 — Variables and Data Types

A **variable** is like a labelled box. You store something inside and refer to it by name whenever you need it.

```
name = "Alice"   →   box labelled 'name', contains the text 'Alice'
age  = 21        →   box labelled 'age',  contains the number 21
```

Python has four basic data types:

| Type | Name | Example |
|------|------|---------|
| Text | `str` | `"Hello"` |
| Whole number | `int` | `21` |
| Decimal number | `float` | `4.5` |
| True / False | `bool` | `True`, `False` |

Unlike other languages, Python infers the type by its value.

````python
# Creating one variable of each type
name    = "Alice"
age     = 21
gpa     = 3.8
active  = True

print(name)
print(age)
print(gpa)
print(active)
````

Use `type()` to ask Python what kind of value a variable holds:

````python
print(type(name))
print(type(age))
print(type(gpa))
print(type(active))
````

### ⚠️ Common errors — read them, do not skip them

Run the cell below and look at what happens. **Do not fix it yet — read the error message carefully.**

````python
# What happens if we add text and a number?
result = "My GPA is: " + 3.8
print(result)
````

The error reads `TypeError: can only concatenate str (not "float") to str`.  
Python does not know how to add text and a number together. You need to **convert** the number to text first using `str()` (a native Python function to convert a value into string):

````python
# Fix: convert the number to a string
result = "My GPA is: " + str(3.8)
print(result)

# Other useful conversions:
print(int("21"))    # text "21"  → integer 21
print(float("3.8")) # text "3.8" → decimal 3.8
````

> **Watch out with booleans:** `True` and `False` are capitalised. Writing `true` or `TRUE` will raise a `NameError`.

````python
# This works:
enrolled = True

# Uncomment the line below to see the error:
enrolled = True
````

## ➕ Block 3 — Operators

### Arithmetic operators

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division (always float) | `5 / 2` | `2.5` |
| `//` | Integer division | `5 // 2` | `2` |
| `%` | Modulo (remainder) | `5 % 2` | `1` |
| `**` | Power | `5 ** 2` | `25` |

````python
# Try each operator
print("Addition:         ", 5 + 3)
print("Subtraction:      ", 5 - 3)
print("Multiplication:   ", 5 * 3)
print("Division:         ", 5 / 2)   # Always returns float
print("Integer division: ", 5 // 2)  # Drops the decimal
print("Modulo (remainder):", 5 % 2)  # How much is left over?
print("Power:            ", 5 ** 2)
````

The `%` operator is very handy for checking whether a number is even or odd:

````python
number = 17
print(number % 2)  # 0 → even, 1 → odd
````

### Comparison operators

These always return `True` or `False`.

| Operator | Meaning |
|----------|---------|
| `==` | Are they equal? |
| `!=` | Are they different? |
| `>` | Greater than? |
| `<` | Less than? |
| `>=` | Greater than or equal? |
| `<=` | Less than or equal? |

> 🚨 **Classic mistake:** `=` **assigns** a value. `==` **compares** two values. They are completely different.

````python
grade = 3.8

print(grade == 3.8)   # Is it exactly 3.8?
print(grade != 4.0)   # Is it different from 4.0?
print(grade >= 2.0)   # Did the student pass? (threshold: 2.0)
print(grade < 4.0)    # Is it below 4.0?
````

## 🖨️ Block 4 — print() and f-strings

You have already used `print()`. Now let's see the **correct and modern** way to embed variables inside text.

````python
name  = "Alice"
grade = 3.8

# Option 1: concatenation with + (works but gets messy)
print("The student " + name + " obtained a GPA of " + str(grade))

# Option 2: f-string (the modern and recommended way)
print(f"The student {name} obtained a GPA of {grade}")
````

**f-strings** are written by placing an `f` before the opening quote. Any variable inside `{}` is inserted automatically.

You can also perform operations directly inside the curly braces:

````python
unit_price = 25
quantity   = 4

print(f"Unit price: ${unit_price}")
print(f"Total ({quantity} units): ${unit_price * quantity}")
print(f"Grade rounded to 2 decimals: {3.16789:.3f}")
#despues de dejar el decimal si quiero redondear el resultado coloco un punto mas la cantidad de decimales que quiero
````

## 🎯 Block 5 — Integrative Exercise

Let's build a **grade report calculator** that uses everything covered today.

### Problem statement
A student has three partial grades. You need to:
1. Calculate the average
2. Determine whether the student passed (average ≥ 2.0)
3. Print a formatted report with the results

Read the code first, then run it, then change the grade values.

````python
# Step 1: Define the grades
grade1 = 1.5
grade2 = 2.0
grade3 = 1.7

print(f"Grade 1: {grade1}")
print(f"Grade 2: {grade2}")
print(f"Grade 3: {grade3}")
````

````python
# Step 2: Calculate the average
average = (grade1 + grade2 + grade3) / 3

print(f"Average: {average:.2f}")
````

````python
# Step 3: Did the student pass?
passed = average >= 2.0

print(f"Did the student pass? {passed}")
````

````python
# Step 4: Full formatted report
student = "Alice Johnson"
course  = "Introduction to Statistics"

print("=" * 45)
print("  ACADEMIC REPORT")
print("=" * 45)
print(f"  Student : {student}")
print(f"  Course  : {course}")
print("-" * 45)
print(f"  Grade 1 : {grade1}")
print(f"  Grade 2 : {grade2}")
print(f"  Grade 3 : {grade3}")
print("-" * 45)
print(f"  Average : {average:.2f}")
print(f"  Passed?  : {passed}")
print("=" * 45)
````
