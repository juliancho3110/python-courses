## 🧠 Block 0 — Programming Thinking: From Problem to Code

Before we write any new Python syntax, we need to address something more important: **how do you turn a sentence like “find the student with the highest grade” into actual working code?**

This is the skill that separates someone who understands syntax from someone who can actually program. It is not obvious. It is not natural. It is a learned process — and this block gives you that process.

### The recipe analogy

Imagine a friend asks you: *“Make a fruit salad.”*

You do not grab a random piece of fruit and start cutting. You think:
1. Which fruits do I have? **(inputs)**
2. What should the salad look like at the end? **(output)**
3. What steps do I follow in order? **(decomposition)**

Programming works exactly the same way.

- The **problem statement** is the request.
- Your **data** (a list, a dictionary) is the raw ingredients.
- The **code** is the sequence of steps that turns ingredients into the finished result.

The skill you are building is learning how to **write that recipe** — before you touch the keyboard.

### The 5-step process

Every time you face a programming problem, follow these five steps in order.

| Step | What to do | Key question |
|------|------------|--------------|
| **1. Read** | Read the whole problem — twice. Do not touch the keyboard yet. | *What is this actually asking me to do?* |
| **2. Identify** | Name your inputs and outputs explicitly. | *What data do I have? What do I need to produce?* |
| **3. Decompose** | Break the problem into 3–5 small steps. | *What must happen first? Then what?* |
| **4. Pseudocode** | Write each step as a sentence, one per line. | *Can I read each step out loud and have it make sense?* |
| **5. Translate** | Convert each pseudocode line to Python, one at a time. | *Which Python tool maps to this sentence?* |

> **Rule:** If you are staring at a blank cell and do not know what to type, you skipped a step. Go back to Step 1 and do not jump ahead.

### Worked example — the full 5-step process in action

**Problem statement:**  
*“You have a list of student grades. Calculate the average and print whether the student passed (average ≥ 3.0) or failed.”*

---

**Step 1 — Read carefully**  
Reading twice reveals two tasks: (1) calculate an average, (2) compare it to 3.0 and print a message.

---

**Step 2 — Identify inputs and outputs**
- **Input:** a list of numbers (grades)
- **Output:** one number (the average) + one string (`"Passed"` or `"Failed"`)

---

**Step 3 — Decompose**
1. Add up all the grades → get a total.
2. Divide the total by the number of grades → get the average.
3. If average ≥ 3.0 → student passed. Otherwise → student failed.
4. Print the result.

---

**Step 4 — Pseudocode**
```
total = 0
for each grade in the list:
    total = total + grade
average = total / number of grades
if average >= 3.0:
    print "Passed"
else:
    print "Failed"
```

---

**Step 5 — Translate** → see the code cell below. Each Python line is labelled with the pseudocode line it came from.

````python
# STEP 5: Translating pseudocode to Python, line by line.
grades = [3.5, 2.8, 4.0, 3.2, 3.9]   # Input: the list of grades.

# Pseudocode: total = 0
total = 0

# Pseudocode: for each grade in the list: total = total + grade
for grade in grades:
    total += grade
# Round 1: grade = 3.5  →  total = 0    + 3.5 = 3.5
# Round 2: grade = 2.8  →  total = 3.5  + 2.8 = 6.3
# Round 3: grade = 4.0  →  total = 6.3  + 4.0 = 10.3
# Round 4: grade = 3.2  →  total = 10.3 + 3.2 = 13.5
# Round 5: grade = 3.9  →  total = 13.5 + 3.9 = 17.4

# Pseudocode: average = total / number of grades
average = total / len(grades)

# Pseudocode: if average >= 3.0 -> print result
if average >= 3.0:
    print(f"Average: {average:.2f} — Passed ✓")
else:
    print(f"Average: {average:.2f} — Failed ✗")
````

### The three fundamental patterns

Almost every programming problem you will encounter in this course follows one of three patterns.  
**Recognizing the pattern before you start typing** is the single most useful habit you can build.

| Pattern | The problem asks you to... | Starting template |
|---------|---------------------------|-------------------|
| **Accumulator** | calculate a total, sum, count, average | `total = 0` → loop → `total += x` |
| **Tracker** | find the highest, lowest, best, worst item | `best = list[0]` → loop → `if x > best: best = x` |
| **Filter** | list all items that match a condition | `result = []` → loop → `if condition: result.append(x)` |

Study the next three cells carefully. You will use these patterns in every exercise.

````python
# PATTERN 1 — ACCUMULATOR
# Problem: "Calculate the total amount spent on travel expenses."
#
# Pattern recognition:
#   "calculate the total" → Accumulator.
#   I have: a list of amounts.  I need: one number (the sum).
#   Template: start at 0, add each amount inside a loop.

travel_expenses = [320000, 85000, 540000, 210000, 95000]  # COP

total = 0                         # Start the accumulator at zero.
for amount in travel_expenses:
    total += amount               # Add each amount to the running total.
# Round 1: amount = 320000  →  total = 0       + 320000 = 320000
# Round 2: amount =  85000  →  total = 320000  +  85000 = 405000
# Round 3: amount = 540000  →  total = 405000  + 540000 = 945000
# Round 4: amount = 210000  →  total = 945000  + 210000 = 1155000
# Round 5: amount =  95000  →  total = 1155000 +  95000 = 1250000

print(f"Total travel expenses: {total:,} COP")
````

````python
# PATTERN 2 — TRACKER
# Problem: "Find the most expensive travel expense."
#
# Pattern recognition:
#   "find the highest" → Tracker.
#   I have: a list of amounts.  I need: one number (the maximum).
#   Template: assume the first is the best, compare every other item against it.

travel_expenses = [320000, 85000, 540000, 210000, 95000]  # COP

highest = travel_expenses[0]      # Assume the first is the highest.
for amount in travel_expenses:
    if amount > highest:
        highest = amount          # Update only when we find something bigger.
# Round 1: amount = 320000  →  320000 > 320000? No  →  highest = 320000
# Round 2: amount =  85000  →   85000 > 320000? No  →  highest = 320000
# Round 3: amount = 540000  →  540000 > 320000? Yes →  highest = 540000
# Round 4: amount = 210000  →  210000 > 540000? No  →  highest = 540000
# Round 5: amount =  95000  →   95000 > 540000? No  →  highest = 540000

print(f"Most expensive trip: {highest:,} COP")
````

````python
# PATTERN 3 — FILTER
# Problem: "List all travel expenses over 200,000 COP."
#
# Pattern recognition:
#   "list all items that..." → Filter.
#   I have: a list of amounts.  I need: a shorter list (only those that qualify).
#   Template: start with an empty list, add items that pass the condition.

travel_expenses = [320000, 85000, 540000, 210000, 95000]  # COP

expensive_trips = []              # Start with an empty result.
for amount in travel_expenses:
    if amount > 200000:
        expensive_trips.append(amount)   # Only add it if it qualifies.
# Round 1: 320000 > 200000? Yes →  expensive_trips = [320000]
# Round 2:  85000 > 200000? No  →  expensive_trips = [320000]
# Round 3: 540000 > 200000? Yes →  expensive_trips = [320000, 540000]
# Round 4: 210000 > 200000? Yes →  expensive_trips = [320000, 540000, 210000]
# Round 5:  95000 > 200000? No  →  expensive_trips = [320000, 540000, 210000]

print(f"Expensive trips: {expensive_trips}")
print(f"Count: {len(expensive_trips)}")
````

## 📋 Block 1 — Lists

A **list** is an ordered collection of items. Items can be of any type, and the list can grow or shrink.

```python
my_list = [item1, item2, item3]
```

Think of it as a column in a spreadsheet — ordered, and each item has a position (index).

````python
# Creating lists.
grades      = [3.5, 2.8, 4.0, 3.2, 3.9]
students    = ["Alice", "Bob", "Carol", "David"]
mixed       = ["Alice", 21, 3.8, True]   # Lists can mix types (though unusual in data).
empty_list  = []                          # An empty list.

print(grades)
print(students)
print(f"Number of grades: {len(grades)}")
````

### Indexing — accessing individual items

Python counts positions starting from **0**, not 1. Negative indices count from the end.

```
grades = [3.5,  2.8,  4.0,  3.2,  3.9]
index:     0     1     2     3     4
negative: -5    -4    -3    -2    -1
```

````python
grades = [3.5, 2.8, 4.0, 3.2, 3.9]

print(grades[0])    # First item.
print(grades[2])    # Third item.
print(grades[-1])   # Last item.
print(grades[-2])   # Second to last.
````

````python
# Read the error before fixing it.
print(grades[10])   # Index does not exist.
````

### Slicing — extracting a portion of a list

`list[start:stop]` — returns items from `start` up to (not including) `stop`.

````python
grades = [3.5, 2.8, 4.0, 3.2, 3.9]

print(grades[1:3])    # Items at index 1 and 2.
print(grades[:3])     # From start up to index 2.
print(grades[2:])     # From index 2 to end.
print(grades[::2])    # Every other item (step = 2).
print(grades[::-1])   # Reversed list.
````

### List methods — modifying a list

A **method** is a function attached to an object. You call it with a dot: `list.method()`.

````python
grades = [3.5, 2.8, 4.0]
print(f"Original:           {grades}")

# append — add one item to the end.
grades.append(3.2)
print(f"After append:       {grades}")

# insert — add at a specific position.
grades.insert(1, 4.5)
print(f"After insert at 1:  {grades}")

# remove — delete the first occurrence of a value.
grades.remove(2.8)
print(f"After remove 2.8:   {grades}")

# pop — remove and return the item at a position (default: last).
last = grades.pop()
print(f"Popped: {last}  →  List: {grades}")

# sort — sort in place.
grades.sort()
print(f"Sorted:             {grades}")

# reverse — reverse in place.
grades.reverse()
print(f"Reversed:           {grades}")
````

````python
grades = [3.5, 2.8, 4.0, 3.2, 3.9]

# Built-in functions that work on lists.
print(f"Length : {len(grades)}")
print(f"Sum    : {sum(grades)}")
print(f"Min    : {min(grades)}")
print(f"Max    : {max(grades)}")
print(f"Average: {sum(grades) / len(grades):.2f}")

# Check membership.
print(f"Is 4.0 in the list? {4.0 in grades}")
print(f"Is 5.0 in the list? {5.0 in grades}")
````

## 🗂️ Block 2 — Dictionaries

A **dictionary** stores data as **key-value pairs**. Instead of a position (index), you access values by a meaningful key.

```python
my_dict = {"key1": value1, "key2": value2}
```

Think of it as a single row in a spreadsheet — each column name is a key, and the cell content is the value.

````python
# A student record.
student = {
    "name"    : "Alice Johnson",
    "age"     : 21,
    "major"   : "Data Science",
    "gpa"     : 3.8,
    "enrolled": True
}

print(student)
print(f"Name: {student['name']}")
print(f"GPA : {student['gpa']}")
````

````python
# Adding and modifying values.
student["semester"] = 4          # Add a new key.
student["gpa"]      = 3.9        # Update an existing key.

print(student)

# Removing a key.
del student["enrolled"]
print(f"After del: {student}")
````

````python
# Read the error — then use .get() to fix it safely.
print(student["scholarship"])   # Key does not exist.
````

````python
# .get() returns None (or a default) instead of crashing.
print(student.get("scholarship"))            # Returns None.
print(student.get("scholarship", "No data")) # Returns default.
````

### Iterating over a dictionary

````python
student = {"name": "Alice", "age": 21, "gpa": 3.8, "semester": 4}

# Iterate over keys.
for key in student:
    print(key)

print("---")

# Iterate over key-value pairs (most common).
for key, value in student.items():
    print(f"  {key:10}: {value}")
````

````python
# Useful dictionary methods.
print(list(student.keys()))    # All keys.
print(list(student.values()))  # All values.
print("name" in student)       # Check if a key exists.
````

## 🔗 Block 3 — Combining Lists and Dictionaries

Real data is almost always a **list of dictionaries** — exactly how a spreadsheet works: multiple rows, each with the same fields.

```
Spreadsheet row  →  dictionary
Spreadsheet col  →  list of values with the same key
Full table       →  list of dictionaries
```

````python
# A small dataset: list of student records.
students = [
    {"name": "Alice",   "gpa": 3.8, "major": "Data Science",  "semester": 4},
    {"name": "Bob",     "gpa": 2.5, "major": "Engineering",    "semester": 2},
    {"name": "Carol",   "gpa": 3.2, "major": "Data Science",   "semester": 6},
    {"name": "David",   "gpa": 1.8, "major": "Business",       "semester": 1},
    {"name": "Eve",     "gpa": 4.0, "major": "Data Science",   "semester": 8},
    {"name": "Frank",   "gpa": 2.9, "major": "Engineering",    "semester": 3},
]

# Access a single row.
print(students[0])

# Access a specific field in a row.
print(students[0]["name"])
print(students[2]["gpa"])
````

````python
# Print all students.
for s in students:
    print(f"  {s['name']:8} | GPA: {s['gpa']} | {s['major']}")
````

````python
# Pattern: FILTER
# Problem: "Print only Data Science students with GPA >= 3.0."
#
# I have: a list of student dicts.  I need: to print a subset.
# Steps: loop over all students -> check two conditions -> print if both are true.

print("Data Science students with GPA >= 3.0:")

for s in students:
    if s["major"] == "Data Science" and s["gpa"] >= 3.0:
        print(f"  {s['name']} — GPA: {s['gpa']}")

# Round 1: Alice | major == DS? Yes | gpa >= 3.0? Yes → print
# Round 2: Bob   | major == DS? No  → skip
# Round 3: Carol | major == DS? Yes | gpa >= 3.0? Yes → print
# Round 4: David | major == DS? No  → skip
# Round 5: Eve   | major == DS? Yes | gpa >= 3.0? Yes → print
# Round 6: Frank | major == DS? No  → skip
````

````python
# Pattern: ACCUMULATOR
# Problem: "Calculate the average GPA across all students."
#
# I have: a list of student dicts, each with a "gpa" key.
# I need: one number (the class average).
# Steps: accumulate total GPA across all students -> divide by count.

total_gpa = 0                    # Start the accumulator at zero.
for s in students:
    total_gpa += s["gpa"]        # Add this student's GPA to the running total.
# Round 1: s["gpa"] = 3.8  →  total_gpa = 0    + 3.8 = 3.8
# Round 2: s["gpa"] = 2.5  →  total_gpa = 3.8  + 2.5 = 6.3
# Round 3: s["gpa"] = 3.2  →  total_gpa = 6.3  + 3.2 = 9.5
# Round 4: s["gpa"] = 1.8  →  total_gpa = 9.5  + 1.8 = 11.3
# Round 5: s["gpa"] = 4.0  →  total_gpa = 11.3 + 4.0 = 15.3
# Round 6: s["gpa"] = 2.9  →  total_gpa = 15.3 + 2.9 = 18.2

average_gpa = total_gpa / len(students)
print(f"Class average GPA: {average_gpa:.2f}")
````

````python
# Pattern: GROUPED ACCUMULATOR (a special variant)
# Problem: "Group students by major — build a dict where each key is a major
#           and the value is a list of student names in that major."
#
# I have: a list of student dicts.  I need: a dictionary of lists.
# Steps:
#   1. Start with an empty dictionary.
#   2. For each student, get their major.
#   3. If we have never seen that major before -> create an empty list for it.
#   4. Append the student's name to the correct major's list.

by_major = {}                    # Start with an empty dictionary.

for s in students:
    major = s["major"]           # Get this student's major.
    if major not in by_major:
        by_major[major] = []     # First time we see this major: create its list.
    by_major[major].append(s["name"])   # Add name to the right major's list.

for major, names in by_major.items():
    print(f"{major}: {names}")
````

## 🟠 Optional · Git/GitHub — Staging and .gitignore

> Skip this section if Git was not activated for your course.

### The staging area

Git has three zones:

```
Working directory  →  Staging area  →  Repository
  (your files)       (git add)        (git commit)
```

You **choose** which changes to include in each commit using `git add`.  
This lets you make two unrelated changes and commit them separately with different messages.

```bash
# Stage a specific file
git add module3_data_structures.ipynb

# Stage everything that changed
git add .

# Check what is staged vs unstaged
git status

# Unstage a file (without losing the changes)
git restore --staged module3_data_structures.ipynb

# Commit what is staged
git commit -m "Add Module 3 notebook — lists and dictionaries"
```

### .gitignore — files Git should never track

Some files should never be committed: temporary files, secrets, large datasets, system files.

Create a file called `.gitignore` in your repo root:

```
# .gitignore

# Jupyter Notebook checkpoints
.ipynb_checkpoints/

# Python cache files
__pycache__/
*.pyc

# Data files you do not want to share
*.csv
*.xlsx

# OS files
.DS_Store
Thumbs.db
```

After saving `.gitignore`, run `git status` — those files will no longer appear.

```bash
# Commit your .gitignore too
git add .gitignore
git commit -m "Add .gitignore for Python and Jupyter files"
```

| Command | What it does |
|---------|-------------|
| `git add .` | Stages all changed files |
| `git restore --staged <f>` | Unstages a file |
| `git diff` | Shows unstaged changes |
| `git diff --staged` | Shows staged changes |
| `.gitignore` | Tells Git which files to ignore |

## 📋 Module 3 Summary

| Concept | Syntax | Key point |
|---------|--------|-----------|
| 5-step process | Read → Identify → Decompose → Pseudocode → Translate | Use it every time you face a blank cell |
| Accumulator | `total = 0` → loop → `total += x` | Building up a result step by step |
| Tracker | `best = list[0]` → loop → `if x > best: best = x` | Finding the maximum or minimum |
| Filter | `result = []` → loop → `if cond: result.append(x)` | Selecting a subset of items |
| Create list | `[1, 2, 3]` | Ordered collection of items |
| Index | `list[0]`, `list[-1]` | Access a single item |
| Slice | `list[1:3]` | Extract a portion |
| Add item | `.append()`, `.insert()` | Growing a list |
| Remove item | `.remove()`, `.pop()` | Shrinking a list |
| List stats | `len()`, `sum()`, `min()`, `max()` | Quick aggregations |
| Create dict | `{"key": value}` | Named fields (like a spreadsheet row) |
| Access value | `d["key"]`, `d.get("key")` | Read a field safely |
| Iterate dict | `for k, v in d.items()` | Loop over all fields |
| List of dicts | `[{...}, {...}]` | Tabular data (like a full spreadsheet) |
