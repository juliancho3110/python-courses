## 🔷 Block 0 — The mental model: a guard at the door


### Analogy: security guard

Imagine a security guard at a building entrance.
His job is simple:

> **While** the visitor does not have a badge — *keep asking for identification.*
> The moment the visitor shows a valid badge — *stop asking and let them in.*

A `while` loop works the same way:
1. **Check** the condition (does the visitor have a badge?).
2. If `True` — run the body (ask for identification).
3. **Go back to step 1** and check again.
4. If `False` — stop and move on.

```
while <condition is True>:
    # body: what happens each round
    # IMPORTANT: something inside the body must eventually make the condition False
```

The **critical rule**: every `while` loop must have a way to eventually make the condition `False`.
If it never becomes `False`, the loop runs forever (infinite loop → your program freezes).


### Reading a while loop round by round

Before you run any loop, practice **tracing it on paper**:
write down the value of every variable at the start of each round.

> **Before running:** predict — how many times will this loop print?


````python
# Countdown from 4 to 1.
counter = 4

while counter > 0:
    print(f"Round {5 - counter}: counter = {counter}")
    counter -= 1   # This line makes the condition eventually False.

print("Loop finished.")

# Round 1: counter = 4  →  4 > 0 is True  →  prints, then counter = 3
# Round 2: counter = 3  →  3 > 0 is True  →  prints, then counter = 2
# Round 3: counter = 2  →  2 > 0 is True  →  prints, then counter = 1
# Round 4: counter = 1  →  1 > 0 is True  →  prints, then counter = 0
# Check:   counter = 0  →  0 > 0 is False →  loop stops.
````

### `for` vs `while` — the same task, two ways

| | `for` | `while` |
|---|---|---|
| **Use when** | You know *how many* iterations in advance | You repeat *until a condition changes* |
| **Counter** | Managed automatically by `range()` | You manage it yourself |
| **Risk** | Off-by-one errors | Infinite loops |

Same task written both ways — read them side by side:


````python
# Task: print numbers 1, 2, 3, 4, 5.

# --- With for ---
print("With for:")
for n in range(1, 6):
    print(n)

# --- With while ---
print("\nWith while:")
n = 1           # Step A: initialise the counter BEFORE the loop.
while n <= 5:   # Step B: condition (loop continues while this is True).
    print(n)
    n += 1      # Step C: update the counter (do NOT forget this!).

# Key insight: while needs THREE things that for handles automatically:
# A. Initialise  B. Condition  C. Update
# If you forget C, you get an infinite loop.
````

## 🔷 Block 1 — The three most common mistakes


### Mistake 1: forgetting to update the counter (infinite loop)

The cell below has a deliberate bug. **Read the code, predict what happens, then run it.**

> ⚠️ **Warning:** this cell will loop forever. Click the ■ Stop button in the toolbar
> (or press **I, I** in command mode) to interrupt the kernel.


````python
# ❌ DELIBERATE BUG — read this before running.
# counter is never updated inside the loop, so counter > 0 is always True.

counter = 5
while counter > 0:
    print(f"counter = {counter}")   # Prints forever: 5, 5, 5, 5 ...
    # counter -= 1   <-- this line is missing!
````

**Fix:** always make sure something inside the loop body changes the variable
that the condition depends on.


````python
# ✅ Fixed version.
counter = 5
while counter > 0:
    print(f"counter = {counter}")
    counter -= 1   # This moves counter toward 0, so the loop will end.

print("Done.")
````

### Mistake 2: wrong initial value (loop never runs)

Sometimes the condition is `False` before the loop even starts — so the body
never executes. This is the *opposite* of an infinite loop, but equally confusing.

> **Before running:** predict — will this loop print anything?


````python
# counter starts at 0, but the condition requires counter > 0.
# The condition is immediately False — the body never runs.
counter = 0
while counter > 0:
    print(f"counter = {counter}")   # This line is never reached.
    counter -= 1

print("Loop ended (or never started).")

# Fix: initialise counter to a value that makes the condition True on round 1.
# In this case, counter = 5 (any positive number).
````

### Mistake 3: off-by-one in the condition (`<` vs `<=`)

> **Before running:** how many numbers does each loop print?


````python
# Version A: strict less-than
print("Version A (while n < 5):")
n = 1
while n < 5:
    print(n)
    n += 1
# Round 1: n=1  1<5 True   print 1  n=2
# Round 2: n=2  2<5 True   print 2  n=3
# Round 3: n=3  3<5 True   print 3  n=4
# Round 4: n=4  4<5 True   print 4  n=5
# Check:   n=5  5<5 False  stop.

# Version B: less-than-or-equal
print("\nVersion B (while n <= 5):")
n = 1
while n <= 5:
    print(n)
    n += 1
# Same trace but n=5 also passes (5<=5 is True) → prints one extra number.
````

## 🔷 Block 2 — Sentinel values: “loop until a stop signal”


### Analogy: the checkout conveyor belt

At a supermarket checkout, the cashier scans items one by one.
They do not know in advance how many items there are.
The **divider bar** is the sentinel: when it appears, the cashier stops scanning.

In code, a **sentinel value** is a special value that signals “stop the loop.”
Common sentinels: `0`, `-1`, `"done"`, `""` (empty string).

**Pattern (Accumulator + sentinel) — reading from a list:**
```python
values = [50000, 120000, 30000, 0]   # 0 is the sentinel (end marker).
i     = 0
total = 0
value = values[i]
i += 1
while value != 0:       # sentinel check
    total += value      # accumulate
    value = values[i]   # read next
    i += 1
print(f"Total: {total}")
```

> In a real program you would replace the list with `input()`.
> Here we use a predefined list so the cell runs without stopping to ask for typing.


### Example: sum monthly expenses until sentinel 0

**Before you code — decomposition:**

- **Pattern:** Accumulator
- **Inputs:** predefined list of expense amounts; last value is `0` (sentinel)
- **Output:** total sum, count, and average
- **Steps:**
  1. Initialise `total = 0`, `count = 0`, index `i = 0`.
  2. Read the first expense from the list.
  3. While expense is not 0: add to total, increment count, read next expense.
  4. Print total and average.

> The list simulates what a user would type one value at a time.
> Change the values in `expense_entries` to test different scenarios.


````python
# Sum monthly expenses until sentinel 0 is reached.
# Pattern: Accumulator.

# Simulated expense entries; 0 marks the end (sentinel).
# Change these values to test different scenarios.
expense_entries = [50_000, 120_000, 30_000, 0]
i = 0

total   = 0   # Initialise accumulator.
count   = 0   # Track how many expenses were entered.

expense = expense_entries[i]   # Read first value.
i += 1

while expense != 0:               # Sentinel check.
    total   += expense            # Accumulate.
    count   += 1                  # Count.
    expense = expense_entries[i]  # Read next value.
    i += 1

print(f"\nTotal expenses: COP {total:,.0f}")
print(f"Number of expenses entered: {count}")
if count > 0:
    print(f"Average expense: COP {total / count:,.0f}")

# Trace (with expense_entries = [50000, 120000, 30000, 0]):
# Before loop: total=0, count=0, expense=50000
# Round 1: 50000 != 0 True   total=50000   count=1   expense=120000
# Round 2: 120000 != 0 True  total=170000  count=2   expense=30000
# Round 3: 30000 != 0 True   total=200000  count=3   expense=0
# Check:   0 != 0 False  stop.
# Output: total=200000, count=3, avg=66666
````

## 🔷 Block 3 — `break` and `continue`


### `break` — the fire alarm

Imagine an office where employees work through a task list.
Normally they finish every task, but if the **fire alarm** goes off, they
**immediately leave the building** — no matter how many tasks remain.

`break` does the same: it **exits the loop immediately**, even if the condition is still `True`.

### `continue` — the conveyor belt filter

On a factory conveyor belt, a sensor checks each item.
If an item is defective, the sensor **pushes it aside and moves to the next item** —
it does not stop the whole belt.

`continue` does the same: it **skips the rest of the current iteration** and goes
back to the condition check.


> **Before running:** how many numbers will this loop print?


````python
# break: exit the loop immediately when a condition is met.
# Scenario: search for the first product with stock below 10.

products = [
    {"name": "Café molido",  "stock": 45},
    {"name": "Leche entera", "stock": 3},    # Low stock!
    {"name": "Pan tajado",   "stock": 20},
    {"name": "Arroz 500g",   "stock": 8},    # Also low, but we stop at the first.
]

i = 0
found = None

while i < len(products):
    product = products[i]
    if product["stock"] < 10:
        found = product
        break          # Exit immediately: do not check the rest.
    i += 1

# Round 1: i=0  products[0]["stock"]=45  45<10 False  i=1
# Round 2: i=1  products[1]["stock"]=3   3<10 True   found=Leche  BREAK
# (products[2] and products[3] are never checked)

if found:
    print(f'Low stock alert: {found["name"]} (stock: {found["stock"]})')
else:
    print("All products have sufficient stock.")
````

> **Before running:** which numbers will this loop print? Which will it skip?


````python
# continue: skip the current iteration and go back to the condition check.
# Scenario: sum only the positive salaries (ignore negative/zero entries).

salaries = [2800000, -500000, 3200000, 0, 4100000, 1950000]
total = 0
i = 0

while i < len(salaries):
    salary = salaries[i]
    i += 1                  # Always update i BEFORE continue, or infinite loop!
    if salary <= 0:
        continue            # Skip this salary: go back to while condition.
    total += salary         # Only reached if salary > 0.

# Round 1: salary=2800000  >0  total=2800000
# Round 2: salary=-500000  <=0 continue (skip total +=)
# Round 3: salary=3200000  >0  total=6000000
# Round 4: salary=0        <=0 continue (skip total +=)
# Round 5: salary=4100000  >0  total=10100000
# Round 6: salary=1950000  >0  total=12050000

print(f"Total valid salaries: COP {total:,.0f}")
````

## 🔷 Block 4 — Accumulator pattern with `while`


You already know the Accumulator pattern from `for` loops.
The same pattern works with `while` — the only difference is that
you manage the counter yourself.

| Step | `for` | `while` |
|------|-------|---------|
| Initialise | `total = 0` | `total = 0` |
| Counter | `range()` handles it | You write `i = 0` and `i += 1` |
| Accumulate | `total += item` | `total += item` |
| Condition | fixed by `range()` | you write `while i < n:` |


### Example: monthly budget tracker

**Before you code — decomposition:**

- **Pattern:** Accumulator + Tracker
- **Inputs:** list of monthly expenses
- **Output:** total, average, and the most expensive month
- **Steps:**
  1. Initialise `total = 0`, `highest = expenses[0]`, `i = 0`.
  2. While `i < len(expenses)`: add expense to total, check if it is the highest, increment `i`.
  3. Compute average, print results.
- **Pseudocode (line numbers correspond to Python lines below):**
  ```
  [1] total ← 0
  [2] highest ← first expense
  [3] i ← 0
  [4] while i < number of expenses:
  [5]     total ← total + expenses[i]
  [6]     if expenses[i] > highest: highest ← expenses[i]
  [7]     i ← i + 1
  [8] average ← total / number of expenses
  [9] print results
  ```


````python
# Monthly budget tracker using while + Accumulator + Tracker.

expenses = [450000, 820000, 310000, 670000, 950000, 510000]   # COP, Jan-Jun.
months   = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

# [1] Initialise accumulators.
total   = 0
highest = expenses[0]   # [2] Assume first is highest.
best_m  = months[0]
i       = 0             # [3] Counter.

while i < len(expenses):        # [4] Condition.
    total += expenses[i]        # [5] Accumulate total.
    if expenses[i] > highest:   # [6] Track highest.
        highest = expenses[i]
        best_m  = months[i]
    i += 1                      # [7] Update counter.

# [8] Compute average.
average = total / len(expenses)

# Round 1: i=0  total=450000    highest=450000  best_m=Jan  i=1
# Round 2: i=1  total=1270000   highest=820000  best_m=Feb  i=2
# Round 3: i=2  total=1580000   highest=820000  best_m=Feb  i=3
# Round 4: i=3  total=2250000   highest=820000  best_m=Feb  i=4
# Round 5: i=4  total=3200000   highest=950000  best_m=May  i=5
# Round 6: i=5  total=3710000   highest=950000  best_m=May  i=6
# Check:   i=6  6 < 6 False  stop.

# [9] Print results.
print(f"Total spent:         COP {total:,.0f}")
print(f"Average per month:   COP {average:,.0f}")
print(f"Most expensive month: {best_m} (COP {highest:,.0f})")
````

## 📋 Summary

| Concept | Key rule | Common mistake |
|---------|----------|----------------|
| `while` loop | Check → body → check → … | Forgetting the update → infinite loop |
| Counter | Initialise before, update inside | Starting at wrong value |
| Sentinel value | Special value that signals stop | Using the wrong comparison operator |
| `break` | Exits the loop immediately | Using `break` when you meant `continue` |
| `continue` | Skips rest of current iteration | Forgetting to update `i` before `continue` |
| Accumulator | `total = 0` → loop → `total += x` | Same as with `for` loops |

**The three questions to ask before every `while` loop:**
1. What is the **initial value** of my loop variable?
2. What is the **condition** that keeps the loop running?
3. What inside the body **changes** the variable so the condition eventually becomes `False`?

If you can answer all three, your loop will not run forever.
