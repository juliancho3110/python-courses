## 🔷 Block 1 — Matplotlib Basics

### Analogy: painting on a canvas

Matplotlib is like drawing on a canvas.

1. You set up the **canvas** → `plt.figure(figsize=(width, height))`
2. You pick up a **brush** and make a mark → `plt.plot()`, `plt.bar()`, `plt.hist()`, `plt.scatter()`
3. You add **labels and decoration** → `plt.title()`, `plt.xlabel()`, `plt.ylabel()`, `plt.legend()`
4. You **hang the painting on the wall** (display it) → `plt.show()`

Every chart follows this same sequence — canvas → drawing → decoration → display.

### Anatomy of a Matplotlib chart

```
plt.figure(figsize=(8, 4))    ← canvas: width=8 inches, height=4 inches
plt.plot(x, y)                ← drawing function
plt.title("My Chart")         ← decoration: title
plt.xlabel("X axis")          ← decoration: x label
plt.ylabel("Y axis")          ← decoration: y label
plt.legend()                  ← decoration: legend (only needed when you use label=)
plt.tight_layout()            ← prevents labels from being cut off
plt.show()                    ← display
```

> **Rule:** Every chart you create must have a **title**, **xlabel**, and **ylabel**. No exceptions.
> An unlabeled chart is useless — no one knows what they are looking at.

### 1.1 Line chart — showing change over time

**When to use:** When your x-axis is time (months, years, days) and you want to show a trend.

> **Before running:** The revenue numbers go up and down. Predict: will the line be smooth or jagged?

````python
# Monthly revenue trend for 2024.
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [45000000, 52000000, 48000000, 61000000, 55000000, 58000000]

plt.figure(figsize=(8, 4))           # Canvas: 8 inches wide, 4 inches tall.
plt.plot(
    months,                          # X values — categories on the horizontal axis.
    revenue,                         # Y values — the numbers to plot.
    marker="+",                      # Draw a dot at each data point.
    color="green",               # Line and dot color.
    linewidth= 1                   # Line thickness.
)
plt.title("Monthly Revenue — 2024")  # Title at the top.
plt.xlabel("Month")                  # Label for the horizontal axis.
plt.ylabel("Revenue (COP)")          # Label for the vertical axis.
plt.tight_layout()                   # Prevent labels from being clipped.
plt.show()                       # Display the chart.
````

**What to notice:**
- `marker="o"` adds a dot at each month — helpful for reading exact values.
- `linewidth=2` makes the line thicker and easier to see.
- Revenue dips in March, peaks in April. You can see the trend immediately — that is the power of visualization.

### 1.2 Bar chart — comparing categories

**When to use:** When you want to compare a numeric value across a set of categories.

Here we use `.groupby()` from Module 7 to compute the average salary per department first.

````python
# Compute average salary per department (groupby from M7).
dept_avg = employees.groupby("department")["salary"].mean().reset_index()
print(dept_avg)

plt.figure(figsize=(6, 4))
plt.bar(
    dept_avg["department"],                              # X: category labels.
    dept_avg["salary"],                                  # Y: bar heights.
    color=["steelblue", "coral", "mediumseagreen"]       # One color per bar.
)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary (COP)")
plt.tight_layout()
plt.show()
````

### 1.3 Histogram — distribution of one variable

**When to use:** When you want to see *how values are spread out* — are most salaries clustered together? Are there outliers?

A histogram groups values into **bins** (ranges). Each bar shows how many values fall into that range.

> **Before running:** There are 12 employees. With `bins=6`, each bin covers roughly `(5500000 - 2800000) / 6 ≈ 450000` COP. Predict which bin will be the tallest.

````python
plt.figure(figsize=(7, 4))
plt.hist(
    employees["salary"],   # The data to distribute into bins.
    bins=6,                # Number of bars (ranges).
    color="steelblue",
    edgecolor="white"      # White border between bars — easier to read.
)
plt.title("Salary Distribution")
plt.xlabel("Salary (COP)")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
````

**About `bins`:**
- More bins → more detail, each bar covers a smaller range.
- Fewer bins → more summary, each bar covers a larger range.
- Start with `bins=6` to `bins=10` and adjust based on what you see.

Try changing `bins=6` to `bins=3` and then `bins=12`. Watch how the shape changes.

### 1.4 Scatter plot — relationship between two variables

**When to use:** When you want to see if two numeric variables move together.
Does salary increase with years of experience? A scatter plot answers this visually.

> **Before running:** Predict — do you expect the dots to go from bottom-left to top-right (positive trend), or will they be scattered randomly?

````python
plt.figure(figsize=(7, 5))
plt.scatter(
    employees["years_exp"],  # X axis — the "cause" variable.
    employees["salary"],     # Y axis — the "effect" variable.
    color="coral",
    s=100,                   # Dot size in points squared.
    alpha=0.7          # Transparency: 0 = invisible, 1 = solid.
)
plt.title("Salary vs Years of Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Salary (COP)")
plt.tight_layout()
plt.show()
````

**What to look for:**
- If dots trend upward → positive relationship (more experience = higher salary).
- If dots are random → no clear relationship.
- `alpha=0.7` is useful when dots overlap — you can see density through transparency.

## 🔷 Block 2 — Seaborn

### Analogy: Matplotlib with a personal stylist

If Matplotlib is drawing on a blank canvas, **Seaborn is Matplotlib with a personal stylist**.

- It makes statistical charts look professional with less code.
- It understands Pandas DataFrames natively — you pass the DataFrame and column names directly.
- It automatically computes statistics for you (means, confidence intervals, distributions).

**Key difference:**

| Matplotlib | Seaborn |
|------------|---------|
| `plt.bar(x_values, y_values)` — you compute first | `sns.barplot(data=df, x="col", y="col")` — it computes for you |
| Lower level, more control | Higher level, less code |
| No confidence intervals | Adds error bars automatically |

Both libraries work together. Seaborn draws on a Matplotlib canvas — you can still use `plt.title()` etc. after a Seaborn call.

---

### 2.1 sns.barplot — with confidence intervals

> **Before running:** Compare this to the `plt.bar()` chart from Block 1. Notice the error bars — what do you think they represent?

````python
plt.figure(figsize=(7, 5))
sns.barplot(
    data=employees,          # Pass the full DataFrame.
    x="department",          # Column name for the x axis.
    y="salary",              # Column name for the y axis (Seaborn computes the mean).
    palette="pastel"         # Color palette name.
)
plt.title("Average Salary by Department (with confidence interval)")
plt.xlabel("Department")
plt.ylabel("Salary (COP)")
plt.tight_layout()
plt.show()
````

**What are the error bars?**
The black lines on top of each bar are **confidence intervals**. They show the spread of values within each group.
- A short bar = all employees in that department earn similar salaries.
- A tall bar = salaries in that department vary widely.
- If error bars from two groups overlap, the difference between them might not be meaningful.

**Seaborn computed the group means automatically** — no need to `groupby()` first.

### 2.2 sns.boxplot — distribution per group

A **box plot** shows the 5-number summary for each group:

```
|--- whisker ---|  [  Q1 | median | Q3  ]  |--- whisker ---|
                   └────── box ──────┘
```

- **Median** (line in the middle): 50% of values are below this.
- **Q1** (left edge of box): 25% of values are below this.
- **Q3** (right edge of box): 75% of values are below this.
- **Whiskers**: extend to the smallest/largest non-outlier values.
- **Dots beyond whiskers**: outliers.

> This connects to Module 9 — Descriptive Statistics, where we will compute these numbers explicitly.

````python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=employees,
    x="department",
    y="salary",
    palette="Set2"
)
plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary (COP)")
plt.tight_layout()
plt.show()
````

**How to read this:**
- Is one department's box higher? → That department pays more on average.
- Is one department's box taller? → Salaries in that department are more spread out.
- Any dots? → Outliers — employees who earn significantly more or less than their colleagues.

### 2.3 sns.histplot — histogram with KDE

Seaborn's `histplot` can overlay a **KDE curve** (Kernel Density Estimate) — a smooth curve that approximates the shape of the distribution.

````python
plt.figure(figsize=(7, 4))
sns.histplot(
    employees["salary"],
    kde=True,            # Overlay a smooth density curve.
    color="steelblue",
    bins=8
)
plt.title("Salary Distribution")
plt.xlabel("Salary (COP)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
````

**The KDE curve** (`kde=True`) smooths the histogram into a curve that is easier to interpret.
The peak of the curve shows the most common salary range.
If the curve has two peaks (bimodal), it suggests two distinct salary groups exist in your data.

### 2.4 sns.scatterplot — with a third variable as color (hue)

Seaborn's scatter plot adds one key feature over `plt.scatter`: the **`hue` parameter**.
It colors each dot by a third variable — so you see three dimensions at once: x, y, and color.

````python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=employees,
    x="years_exp",
    y="salary",
    hue="department",    # Color each dot by department.
    s=100                # Dot size.
)
plt.title("Salary vs Experience (colored by Department)")
plt.xlabel("Years of Experience")
plt.ylabel("Salary (COP)")
plt.tight_layout()
plt.show()
````

**What to look for:**
- Do dots of the same color (department) cluster together?
- Does one department's dots consistently appear higher on the y-axis?
- The legend is created automatically because of `hue=` — Seaborn handles this for you.

### 2.5 sns.heatmap — correlation matrix

A **correlation matrix** shows the relationship between every pair of numeric columns.
Values range from **-1 to +1**:

| Value | Meaning |
|-------|---------|
| Close to **+1** | Both variables increase together (positive correlation) |
| Close to **-1** | One increases while the other decreases (negative correlation) |
| Close to **0** | No clear relationship |

> **Important:** Correlation does NOT mean causation. Two variables can move together by coincidence or because of a third hidden factor.

````python
# Compute the correlation matrix for numeric columns.
corr = employees[["salary", "years_exp"]].corr()
print("Correlation matrix:")
print(corr)

plt.figure(figsize=(5, 4))
sns.heatmap(
    corr,
    annot=True,          # Show the numeric values inside each cell.
    fmt=".2f",           # Format: 2 decimal places.
    cmap="coolwarm",     # Color map: blue = negative, red = positive.
    vmin=-1,             # Fix the color scale from -1 to +1.
    vmax=1
)
plt.title("Correlation Matrix — salary vs years_exp")
plt.tight_layout()
plt.show()
````

**Reading the heatmap:**
- The diagonal is always 1.0 (a variable is perfectly correlated with itself).
- The off-diagonal value tells you how strongly salary and years_exp are related.
- A warm red means positive correlation; cool blue means negative correlation.

## 🔷 Block 3 — Best Practices

### 3.1 Which chart should I use?

| Chart type | Use when | Example question |
|------------|----------|-----------------|
| **Line chart** | Showing change over time | How did revenue evolve month by month? |
| **Bar chart** | Comparing a value across categories | Which department has the highest average salary? |
| **Histogram** | Showing distribution of one variable | How are salaries spread across all employees? |
| **Scatter plot** | Relationship between two numeric variables | Do employees with more experience earn more? |
| **Box plot** | Comparing distributions across groups | How do salary spreads differ by city? |
| **Heatmap** | Matrix of values (correlations, counts) | Which variables are correlated with each other? |

**Quick decision tree:**
- One variable, numeric → **Histogram**
- One variable, categorical → **Bar chart** (counts)
- Two numeric variables → **Scatter plot**
- One numeric per category group → **Bar chart** or **Box plot**
- Trend over time → **Line chart**
- Relationships between many variables → **Heatmap**

---

### 3.2 Non-negotiable rules

These rules apply to every chart you create — in this course, in university assignments, and in professional work.

**1. Always add title, xlabel, ylabel**
```python
plt.title("Descriptive title — what is this chart showing?")
plt.xlabel("What the x-axis represents (units)")
plt.ylabel("What the y-axis represents (units)")
```

**2. Always call `plt.tight_layout()`**
Without it, labels can be cut off at the edges of the figure.

**3. Use readable color palettes**
- Avoid rainbow gradients — they distort perception.
- Good Matplotlib colors: `"steelblue"`, `"coral"`, `"mediumseagreen"`, `"slategray"`
- Good Seaborn palettes: `"pastel"`, `"Set2"`, `"muted"`, `"coolwarm"` (for heatmaps)

**4. Match chart type to data type**
Never use a line chart for unordered categories — the "trend" would be meaningless.

---

### 3.3 Subplots — multiple charts in one figure

`plt.subplots(rows, cols)` creates a grid of charts.
It returns two objects:
- `fig` — the overall figure (the canvas)
- `axes` — a grid of individual chart areas

Each individual chart is accessed as `axes[row, col]` (for a 2D grid) or `axes[i]` (for a 1-row grid).

> **Key difference:** Inside a subplot, use `axes[i].set_title()`, `axes[i].set_xlabel()` instead of `plt.title()`, `plt.xlabel()`.

````python
# Compute the data needed for both charts.
dept_avg = employees.groupby("department")["salary"].mean().reset_index()

# Create a figure with 1 row and 2 columns.
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Left chart — axes[0].
axes[0].bar(dept_avg["department"], dept_avg["salary"], color=["steelblue", "coral", "mediumseagreen"])
axes[0].set_title("Average Salary by Department")
axes[0].set_xlabel("Department")
axes[0].set_ylabel("Average Salary (COP)")

# Right chart — axes[1].
axes[1].hist(employees["salary"], bins=6, color="steelblue", edgecolor="white")
axes[1].set_title("Salary Distribution")
axes[1].set_xlabel("Salary (COP)")
axes[1].set_ylabel("Number of Employees")

plt.tight_layout()   # Adjusts spacing between subplots automatically.
plt.show()
````

**Summary of the subplot pattern:**

```python
fig, axes = plt.subplots(rows, cols, figsize=(width, height))
# Access each cell: axes[row, col] for 2D grid, axes[i] for 1-row grid
axes[0].bar(...)           # draw on the first chart
axes[0].set_title(...)     # decorate the first chart
axes[1].hist(...)          # draw on the second chart
axes[1].set_title(...)     # decorate the second chart
plt.tight_layout()
plt.show()
```

For a 2×2 grid: `axes[0, 0]`, `axes[0, 1]`, `axes[1, 0]`, `axes[1, 1]`.

## 🔷 Block 4 — Mini-project: Employee Analytics Dashboard

### Goal

Build a 2×2 dashboard — four charts, all on one figure — that gives a complete visual overview of the employees dataset.

This is the kind of output you would present to a manager or include in a data analysis report.

### Before you code — full decomposition

**Inputs:** `employees` DataFrame — columns: `name`, `department`, `salary`, `years_exp`, `city`

**Output:** One figure with 4 charts arranged in a 2×2 grid.

**Chart plan (plain English — no Python syntax yet):**

| Position | Chart type | What it shows |
|----------|-----------|---------------|
| Top-left `[0,0]` | Bar chart | Average salary per department |
| Top-right `[0,1]` | Scatter plot | Salary vs years of experience, colored by department |
| Bottom-left `[1,0]` | Box plot | Salary distribution per city |
| Bottom-right `[1,1]` | Bar chart | Number of employees (headcount) per department |

**Steps:**
1. Create a 2×2 figure with `plt.subplots(2, 2)` and a title with `fig.suptitle()`.
2. Compute average salary per department (groupby + mean).
3. Draw chart 1 on `axes[0, 0]` — bar chart.
4. Draw chart 2 on `axes[0, 1]` — scatter plot using Seaborn with hue.
5. Compute headcount per department (groupby + count or value_counts).
6. Draw chart 3 on `axes[1, 0]` — boxplot using Seaborn.
7. Draw chart 4 on `axes[1, 1]` — bar chart of headcount.
8. Call tight_layout and show.

**Patterns used:**
- Accumulator (groupby mean and count for charts 1 and 4)
- Relationship visualization (scatter for chart 2)
- Distribution comparison (boxplot for chart 3)

````python
# Precompute groupby results.
dept_avg = employees.groupby("department")["salary"].mean().reset_index()
dept_count = employees.groupby("department")["name"].count().reset_index()
dept_count.columns = ["department", "headcount"]

# Create the 2x2 dashboard figure.
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("Employee Analytics Dashboard", fontsize=16, fontweight="bold")

# ── Chart 1 (top-left) — Average salary per department ──────────────────
colors_dept = ["steelblue", "coral", "mediumseagreen"]
axes[0, 0].bar(dept_avg["department"], dept_avg["salary"], color=colors_dept)
axes[0, 0].set_title("Average Salary by Department")
axes[0, 0].set_xlabel("Department")
axes[0, 0].set_ylabel("Average Salary (COP)")

# ── Chart 2 (top-right) — Salary vs experience, colored by department ───
sns.scatterplot(
    data=employees,
    x="years_exp",
    y="salary",
    hue="department",
    s=100,
    ax=axes[0, 1]         # Pass ax= to tell Seaborn which subplot to draw on.
)
axes[0, 1].set_title("Salary vs Years of Experience")
axes[0, 1].set_xlabel("Years of Experience")
axes[0, 1].set_ylabel("Salary (COP)")

# ── Chart 3 (bottom-left) — Salary distribution per city ────────────────
sns.boxplot(
    data=employees,
    x="city",
    y="salary",
    palette="Set2",
    ax=axes[1, 0]
)
axes[1, 0].set_title("Salary Distribution by City")
axes[1, 0].set_xlabel("City")
axes[1, 0].set_ylabel("Salary (COP)")

# ── Chart 4 (bottom-right) — Headcount per department ───────────────────
axes[1, 1].bar(dept_count["department"], dept_count["headcount"], color=colors_dept)
axes[1, 1].set_title("Headcount by Department")
axes[1, 1].set_xlabel("Department")
axes[1, 1].set_ylabel("Number of Employees")

plt.tight_layout()
plt.show()
````

### Reflection

- **Chart 1** tells us who earns more on average.
- **Chart 2** tells us whether experience predicts salary, and whether departments have different patterns.
- **Chart 3** tells us whether location affects salary spread.
- **Chart 4** tells us the team composition.

Together, these four charts give a much richer picture than any table of numbers. This is the value of visualization.

## 📋 Module 8 Summary

| Concept | Syntax | Key point |
|---------|--------|-----------|
| Line chart | `plt.plot(x, y, marker="o")` | Best for time trends |
| Bar chart | `plt.bar(x, y, color=...)` | Comparing categories |
| Histogram | `plt.hist(data, bins=N)` | Distribution of one variable |
| Scatter plot | `plt.scatter(x, y, alpha=0.7)` | Relationship between two variables |
| Seaborn bar | `sns.barplot(data=df, x=, y=, palette=)` | Adds confidence interval automatically |
| Seaborn box | `sns.boxplot(data=df, x=, y=)` | Distribution per group (5-number summary) |
| Seaborn histplot | `sns.histplot(data, kde=True)` | Histogram + smooth density curve |
| Seaborn scatter | `sns.scatterplot(data=df, x=, y=, hue=)` | Third variable as color |
| Heatmap | `sns.heatmap(df.corr(), annot=True, fmt=".2f")` | Correlation matrix |
| Subplots | `fig, axes = plt.subplots(r, c, figsize=(...))` | Multiple charts in one figure |
| Subplot decoration | `axes[i].set_title()` / `axes[i, j].set_xlabel()` | Use `.set_` prefix inside subplots |
| Mandatory trio | `plt.title()` + `plt.xlabel()` + `plt.ylabel()` | Never omit — unlabeled charts are useless |
| Layout fix | `plt.tight_layout()` | Prevents labels from being cut off |
