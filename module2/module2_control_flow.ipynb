## 🤔 Block 1 — Conditionals: if, elif, else (40 min)

So far your code runs **top to bottom, always**. Conditionals let Python **make decisions** — run some lines only when a condition is true.

The basic structure is:

```python
if condition:
    # runs only when condition is True
else:
    # runs only when condition is False
```

> 🚨 **Indentation is not optional.** Python uses spaces (4 spaces or 1 tab) to know which lines belong inside an `if`. A missing indent is a `SyntaxError`.

````python
# Basic if/else
grade = 3.5

if grade >= 2.0:
    print("Passed ✅")
else:
    print("Failed ❌")
````

### elif — more than two options

When you need to check **multiple conditions in sequence**, use `elif` (short for *else if*).
Python checks each condition from top to bottom and runs the **first one that is true** — then skips the rest.

````python
# Letter grade classifier
grade = 3.7

if grade >= 3.7:
    letter = "A"
elif grade >= 3.3:
    letter = "B+"
elif grade >= 3.0:
    letter = "B"
elif grade >= 2.7:
    letter = "C+"
elif grade >= 2.0:
    letter = "C"
else:
    letter = "F"

print(f"Grade: {grade} → Letter: {letter}")
````

Test with several values: `4.0`, `3.5`, `2.5`, `1.0`. Predict the output before running each one.

### Logical operators: and, or, not

These let you combine multiple conditions into one.

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `and` | Both must be True | `age >= 18 and has_id` | True only if both |
| `or` | At least one must be True | `is_admin or is_owner` | True if either |
| `not` | Flips True to False | `not is_banned` | Inverts the value |

````python
age        = 20
has_id     = True
is_student = True

# and: both conditions must be True
print(age >= 18 and has_id)      # Can this person enter?

# or: at least one condition must be True
print(age < 18 or is_student)    # Gets a discount?

# not: inverts the condition
print(not is_student)            # Is NOT a student?
````

````python
# Combining everything: scholarship eligibility
gpa          = 3.8
is_enrolled  = True
has_debt     = False

eligible = gpa >= 3.5 and is_enrolled and not has_debt

if eligible:
    print("Eligible for scholarship ✅")
else:
    print("Not eligible ❌")
````

## 🔁 Block 2 — for loops

A `for` loop repeats a block of code **for each item in a list**.

Before looking at the syntax, let's start from something you already know.

### Step 1 — The long way (you already understand this)

````python
# Printing three numbers the long way
number = 10
print(number)   # number is 10

number = 20
print(number)   # number is 20

number = 30
print(number)   # number is 30
````

That works — but imagine doing it for 1000 numbers.

The `for` loop does **exactly the same thing**, automatically:

```
for   [variable]   in   [list]  :
         ↓                ↓
      the box        where Python takes values from

    [instruction using the variable]
```

````python
# The for loop version — identical result
for number in [10, 20, 30]:
    print(number)

# Round 1: Python puts 10 into 'number'  → prints 10
# Round 2: Python puts 20 into 'number'  → prints 20
# Round 3: Python puts 30 into 'number'  → prints 30
````

### Step 2 — Looping directly over a list of names

````python
names = ["Alice", "Bob", "Carol"]

for name in names:
    print(name)

# Round 1: name = "Alice"  → prints "Alice"
# Round 2: name = "Bob"    → prints "Bob"
# Round 3: name = "Carol"  → prints "Carol"
````

Notice: no numbers, no positions — Python goes through each item directly.

You can do anything with the variable inside the loop:

````python
names = ["Alice", "Bob", "Carol"]

for name in names:
    print(f"Hello, {name}! Welcome to the course.")

# Round 1: name = "Alice"  → prints "Hello, Alice! Welcome to the course."
# Round 2: name = "Bob"    → prints "Hello, Bob! Welcome to the course."
# Round 3: name = "Carol"  → prints "Hello, Carol! Welcome to the course."
````

### Step 3 — Using an index to connect two lists

Sometimes you have two lists where position matters — the first name matches the first grade, the second name matches the second grade, etc.

To connect them you need an **index** — a number that points to the same position in both lists.

First, remember how indexing works:

```
names  = ["Alice",  "Bob",   "Carol" ]
grades = [  3.8,     2.5,     3.2   ]
index:      0        1        2
```

`names[0]` and `grades[0]` both belong to Alice.  
`names[1]` and `grades[1]` both belong to Bob.

````python
names  = ["Alice", "Bob", "Carol"]
grades = [3.8, 2.5, 3.2]

# The long way — you already understand this
print(names[0], grades[0])   # Alice  3.8
print(names[1], grades[1])   # Bob    2.5
print(names[2], grades[2])   # Carol  3.2
````

````python
names  = ["Alice", "Bob", "Carol"]
grades = [3.8, 2.5, 3.2]

# The for loop version — i is the index that moves through both lists
for i in [0, 1, 2]:
    print(names[i], grades[i])

# Round 1: i = 0  →  names[0] = "Alice"  grades[0] = 3.8  →  prints "Alice 3.8"
# Round 2: i = 1  →  names[1] = "Bob"    grades[1] = 2.5  →  prints "Bob 2.5"
# Round 3: i = 2  →  names[2] = "Carol"  grades[2] = 3.2  →  prints "Carol 3.2"
````

### Step 4 — The accumulator pattern

One of the most common patterns: **start with zero, add something on every round**.

Imagine counting money in a piggy bank — you start at $0 and add each coin one by one.

````python
grades = [3.5, 2.8, 4.0]

# The long way (complete to add all grades)
total = 0
total = total + grades[0]   # total is now 3.5
total= total  + grades[1]
total= total  + grades[2]


print(total)
````

````python
grades = [3.5, 2.8, 4.0]

total = 0   # Start at zero — the piggy bank is empty

for grade in grades:
    total = total + grade   # Add each grade to the total

# Round 1: grade = 3.5  →  total = 0   + 3.5 = 3.5
# Round 2: grade = 2.8  →  total = 3.5 + 2.8 = 6.3
# Round 3: grade = 4.0  →  total = 6.3 + 4.0 = 10.3

print(f"Total : {total}")
print(f"Average: {total / len(grades):.2f}")
````

![gif_accumulator](gif_accumulator.gif "gif_accumulator")

`total = total + grade` is so common that Python has a shorthand for it:

````python
grades = [3.5, 2.8, 4.0]
total  = 0

for grade in grades:
    total += grade   # Same as: total = total + grade

# Round 1: grade = 3.5  →  total = 0   + 3.5 = 3.5
# Round 2: grade = 2.8  →  total = 3.5 + 2.8 = 6.3
# Round 3: grade = 4.0  →  total = 6.3 + 4.0 = 10.3

print(total)
````

### Step 5 — Combining for and if

Inside a loop you can add conditions — Python checks the `if` on every round.

````python
names  = ["Alice", "Bob", "Carol", "David"]
grades = [3.8, 2.5, 3.2, 1.9]

for i in [0, 1, 2, 3]:
    if grades[i] >= 3.0:
        print(f"{names[i]} passed")
    else:
        print(f"{names[i]} failed")

# Round 1: i = 0  →  grades[0] = 3.8  →  3.8 >= 3.0  →  "Alice passed"
# Round 2: i = 1  →  grades[1] = 2.5  →  2.5 >= 3.0  →  "Bob failed"
# Round 3: i = 2  →  grades[2] = 3.2  →  3.2 >= 3.0  →  "Carol passed"
# Round 4: i = 3  →  grades[3] = 1.9  →  1.9 >= 3.0  →  "David failed"
````

### Step 5 — range(): stop writing [0, 1, 2] by hand

Earlier you wrote `for i in [0, 1, 2]`. That works — but what if you have 100 items?  
`range(n)` generates that list automatically.

````python
# These two are identical
for i in [0, 1, 2]:
    print(i)

print("---")

for i in range(3):   # range(3) generates [0, 1, 2]
    print(i)
````

````python
# Now crossed lists without writing [0, 1, 2] by hand
names  = ["Alice", "Bob", "Carol"]
grades = [3.8, 2.5, 3.2]

for i in range(3):   # range(3) → [0, 1, 2]
    print(f"{names[i]} has a grade of {grades[i]}")

# Round 1: i = 0  →  names[0] = "Alice"  grades[0] = 3.8
# Round 2: i = 1  →  names[1] = "Bob"    grades[1] = 2.5
# Round 3: i = 2  →  names[2] = "Carol"  grades[2] = 3.2
````

````python
# Even better: use len() so it works for any list size
names  = ["Alice", "Bob", "Carol", "David", "Eve"]
grades = [3.8, 2.5, 3.2, 1.9, 4.0]

for i in range(len(names)):   # range(5) → [0, 1, 2, 3, 4]
    print(f"{names[i]}: {grades[i]}")
````

### Step 6 — enumerate(): index and value at the same time

`enumerate()` gives you both the index and the value in one step — no need to write `names[i]` separately.

````python
names = ["Alice", "Bob", "Carol"]

# Without enumerate
for i in range(len(names)):
    print(i, names[i])

print("---")

# With enumerate — cleaner, same result
for i, name in enumerate(names):
    print(i, name)

# Round 1: i = 0, name = "Alice"
# Round 2: i = 1, name = "Bob"
# Round 3: i = 2, name = "Carol"
````

## ⏳ Block 3 — while loops (35 min)

A `while` loop repeats **as long as a condition remains True**.  
Use it when you **do not know in advance** how many iterations you need.

```python
while condition:
    # code that repeats
    # always update something so the condition eventually becomes False
```

> 🚨 **Infinite loop warning:** if the condition never becomes `False`, the loop runs forever. If that happens, click **Kernel → Interrupt** in Jupyter.

````python
# Count down from 5
counter = 5

while counter > 0:
    print(f"Counter: {counter}")
    counter -= 1    # Short for: counter = counter - 1

print("Done!")
````

### When to use for vs while

| Situation | Use |
|-----------|-----|
| You know how many iterations | `for` |
| You are looping over a list | `for` |
| You repeat until something changes | `while` |
| You are waiting for user input | `while` |

````python
# Simulate compound interest until savings double
balance      = 1000.0
annual_rate  = 0.08    # 8% annual interest
target       = balance * 2
year         = 0

while balance < target:
    balance += balance * annual_rate
    year    += 1

print(f"It takes {year} years to double your savings.")
print(f"Final balance: ${balance:,.2f}")
````

````python
# break — exit the loop immediately
# continue — skip to the next iteration

grades = [3.5, 2.8, -1, 4.0, 3.2]   # -1 means no grade recorded

for g in grades:
    if g == -1:
        print("Missing grade found — stopping.")
        break              # Exit the loop entirely
    if g < 2.0:
        continue           # Skip failed grades, go to next
    print(f"Valid grade: {g}")
````

### Problem 1 — Sales report

You have daily sales figures for two weeks. Write a program that:
1. Calculates total and average daily sales
2. Counts days above and below average
3. Identifies the best and worst sales day
4. Prints a summary report

````python
# Try solving it here before looking at the guided solution
sales = [320, 410, 290, 510, 380, 450, 270,
         490, 330, 420, 360, 500, 310, 440]

# Write your code here
````

````python
# Guided solution — run only after you have tried
sales = [320, 410, 290, 510, 380, 450, 270,
         490, 330, 420, 360, 500, 310, 440]

# Step 1: Total and average
total   = 0
best    = sales[0]    # Start assuming first day is best
worst   = sales[0]    # And also worst

for s in sales:
    total += s
    if s > best:
        best = s
    if s < worst:
        worst = s

average = total / len(sales)

# Step 2: Count days above and below average
above = 0
below = 0

for s in sales:
    if s >= average:
        above += 1
    else:
        below += 1

# Step 3: Report
print("=" * 40)
print("  TWO-WEEK SALES REPORT")
print("=" * 40)
print(f"  Total sales   : ${total:,}")
print(f"  Daily average : ${average:,.1f}")
print(f"  Best day      : ${best:,}")
print(f"  Worst day     : ${worst:,}")
print(f"  Days above avg: {above}")
print(f"  Days below avg: {below}")
print("=" * 40)
````
