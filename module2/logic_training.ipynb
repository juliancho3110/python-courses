### Exercise 6 — Factorial
*Source: PYnative — Loop Exercise #13*


Calculate and print the factorial of `n = 6`.

The factorial of a number is the product of all integers from 1 to that number:  
`6! = 6 × 5 × 4 × 3 × 2 × 1 = 720`

**Hint:** this is an accumulator — but instead of adding, you multiply.  
Start with `factorial = 1` (not 0 — why?).

**Expected output:**
```
6! = 720
```

````python
n = 6

# Write your solution here
#describiendo el problema: estan requiriendo la impresion n=6 factorial 
#entradas del problema: la entrada del problema es justamente n=6 factorial lo cual es una variable de tipo entero
#salidas del problema: la salida justamente es la impresion la cual es el resultado esperado sea el factorial 

#descomponiendo el programa en pasos: 
#paso 1: asignar un valor a la variable n (en este caso es 6)
#paso 2:asignar el resultado (factorial) el valor de 1
#paso 3:definiendo los valores de 1 hasta n (en este caso especifico hasta 6)
#paso 4: recorrer 1 a 1 los valores
#paso 5: aplicar la formula del factorial (la cual es una multipllicacion entre el valor recorrido y el resultado)
#paso 6: imprimir en formato f string  el resultado del factorial 

#haciendo el pseudocodigo:
#variable n=6
#definiendo variable: f=1
#generar rango de 1-6
#para cada valor generado:(for con valor m)
    #m*f
#imprimir el factorial (f)

#traduciendo el pseudocodigo a codigo:
n=6
f=1
for m in range(1,n+1):
    f=m*f
print(f"{n}!= {f}")



````

### Exercise 8 — Running total
*Source: Simple Programming Problems — Lists #5*


Given the list of sales below, print the **running total** after each sale —  
that is, the accumulated sum after adding each new value.

**Expected output:**
```
After sale 1: 120000
After sale 2: 275000
After sale 3: 390000
After sale 4: 590000
After sale 5: 720000
```

````python
sales = [120000, 155000, 115000, 200000, 130000]   # COP

# Write your solution here
#descripcion del problema: dada la lista en pantalla imprimir el resultado que se genera despues de cada venta la cual es la suma acumulada al agregar un nuevo valor
#entradas del problema: la entrada del problema es la variable lista la cual esta contenida en una lista y sus  tipo de variables son enteros 
#salidas del problema : la impresion del resultado visto en pantalla despues de calcular las sumas acumuladas al agregar un nuevo valor

#descomponiendo el programa paso a paso
#paso 1: asignar una variable que contenga la lista dada anteriormente
#paso 2: asignar una variable que contenga la sumatoria de la lista
#paso 3: hacer la sumatoria de la lista anterior despues de que se añada un valor nuevo
#paso 4: imprimir el resultado esperado

#desarrollando el pseudocodigo:
#sales=[120000, 155000, 115000, 200000, 130000]
#sum= valor inicial 0
#a traves de cada numero hacer el acumulado empezado en la variable anterior (for)
    #asigno una variable s para reemplazar sum en la lista mas la misma variable
#imprimo el resultado de la variable en formato f string

#traduciendo el pseudocodigo a codigo
sales=[120000, 155000, 115000, 200000, 130000]
acum=0
count=1
for s in sales:
    acum=acum+s
    print(f"After sale {count}: {acum}")
    count=count+1
    


````

### Exercise 9 — Largest element in a list
*Source: Simple Programming Problems — Lists #1*


Given the list of grades below, find and print the highest grade **without using `max()`**.  
Use an accumulator that keeps track of the largest value seen so far.

**Hint:** start with `largest = grades[0]` and update it inside the loop.

**Expected output:**
```
Highest grade: 4.7
```

````python
grades = [3.2, 4.1, 2.8, 4.7, 3.9, 1.5, 4.3]

# Write your solution here
#descripcion del problema:
#entrada del problema : la lista de las notas y el tipo de variable de la entrada es de tipo lista de decimales
#salida del problema: imprimir la nota mas alta que seria 4.7 tipo de variable decimal 

#descomponiendo el programa:
#paso 1: definir la variable que almacenara el resultado que comprende la nota mas alta
#paso 2: recorrer la lista de notas 1 a 1 
#paso 3: comparar la variable con la nota que estamos recorriendo. El objetivo es buscar la nota mas grande entre estas 2 
#paso 4: en el caso que la nota sea mayor a la variable, esta debera reemplazarla
#paso 5: hacer todo lo anterior evaluando todas las notas
#paso 6: cumplir con la salida del problema 
#paso 7: imprimir la nota mas alta 

#realizando el pseudocodigo:
#definir variable : largest con valor inicial en la posicion 0 de la lista de notas
#recorrer cada nota de la lista (grades): (for)
    #comparar si largest es mayor a la nota
        #si(if) el resultado es True no se debe modificar largest
    #en otro caso (else )(si largest no es mayor a la nota largest es menor o igual a la nota)
        # modificamos el valor de largest con la nota.
#imprimir (largest en formato f string)

#traduciendo el pseudocodigo a codigo

largest= grades[0]
for g in grades:
    if largest>g:
        largest=largest
    else:
        largest=g
print(f"Highest grade {largest}")




````

```

```
