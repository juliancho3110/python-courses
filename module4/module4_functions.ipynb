## 🔧 Block 1 — Defining and Calling Functions

A **function** is a named, reusable block of code. You define it once and call it as many times as you need.

Without functions, you repeat the same logic everywhere. When something changes, you have to fix it in ten places. With functions, you fix it in one.

```python
def function_name(parameters):
    # code block
    return result
```

> 🚨 **Indentation again:** everything inside a function must be indented. Same rule as `if` and `for`.

````python
# The simplest possible function
def greet():
    print("Hello, student!")

# Defining a function does nothing by itself — you must call it
greet()
greet()
greet()
````

````python
# A function that takes a parameter
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
greet("Bob")
greet("Carol")

#print("hello Alice")
#print("hello Bob")
#print("hello Carol")
#definir funciones me beneficia para reutilizar codigo 
````

### return — getting a value back

A function that only `print()`s is limited — you cannot use its result elsewhere.  
`return` sends a value back to whoever called the function.

````python
# Without return — result is lost
def calculate_average_bad(grades):
    print(sum(grades) / len(grades))   # Prints but does not return

result = calculate_average_bad([3.5, 2.8, 4.0])
print(f"Stored result: {result}")      # None — nothing was returned
````

````python
# With return — result is usable
def calculate_average(grades):
    return sum(grades) / len(grades)

avg = calculate_average([3.5, 2.8, 4.0])
print(f"Average: {avg:.2f}")

# Now you can use the result in other expressions
if calculate_average([3.5, 2.8, 4.0]) >= 2.0:
    print("Passed ✅")
````

````python
# return immediately exits the function — code after it does not run
def check_grade(grade):
    if grade < 0 or grade > 4.0:
        return "Invalid grade"
    if grade >= 2.0:
        return "Passed"
    return "Failed"

print(check_grade(3.5))
print(check_grade(1.8))
print(check_grade(5.0))
````

## ⚙️ Block 2 — Parameters, Defaults, and Multiple Returns

### Default parameter values

You can assign a default value to a parameter. If the caller does not provide it, the default is used.

````python
def calculate_average(grades, round_to=2):
    avg = sum(grades) / len(grades)
    return round(avg, round_to)

grades = [3.5, 2.8, 4.0, 3.123456]

print(calculate_average(grades))        # Uses default: 2 decimal places
print(calculate_average(grades, 4))     # Overrides: 4 decimal places
print(calculate_average(grades, 0))     # Overrides: no decimal places
````

````python
# Keyword arguments — call by name, order does not matter
def describe_student(name, major, semester=1, gpa=0.0):
    return f"{name} | {major} | Semester {semester} | GPA: {gpa}"

# Positional
print(describe_student("Alice", "Data Science", 4, 3.8))

# Keyword — clearer and order-independent
print(describe_student(name="Bob", major="Engineering", gpa=2.9, semester=2))

# Mix: positional first, then keyword
print(describe_student("Carol", "Business", gpa=3.1))
````

### Returning multiple values

Python functions can return more than one value using a **tuple**.

````python
def grade_stats(grades):
    average = sum(grades) / len(grades)
    highest = max(grades)
    lowest  = min(grades)
    passing = sum(1 for g in grades if g >= 2.0)
    return average, highest, lowest, passing   # Returns a tuple

grades = [3.5, 2.8, 4.0, 1.7, 3.2, 2.1]

# Unpack all four values at once
avg, high, low, passed = grade_stats(grades)

print(f"Average: {avg:.2f}")
print(f"Highest: {high}")
print(f"Lowest : {low}")
print(f"Passing: {passed} / {len(grades)}")
````

## 🔭 Block 3 — Variable Scope

**Scope** determines where a variable is visible. This is a common source of bugs.

- **Local variable:** defined inside a function — only visible there
- **Global variable:** defined outside any function — visible everywhere

````python
# Local variable — does not exist outside the function
def calculate():
    result = 42    # Local variable
    print(f"Inside function: {result}")

calculate()

# Read the error — result does not exist out here
print(result)
````

````python
# Global variable — visible everywhere
passing_threshold = 2.0   # Defined globally

def check_grade(grade):
    if grade >= passing_threshold:   # Can read it from inside
        return "Passed"
    return "Failed"

print(check_grade(3.5))
print(check_grade(1.8))
````

````python
# A local variable with the same name as a global does NOT change the global
value = 100

def modify():
    value = 999    # This creates a NEW local variable, does not touch the global
    print(f"Inside:  {value}")

modify()
print(f"Outside: {value}")   # Still 100
````

> **Rule of thumb:** avoid modifying global variables inside functions. Instead, pass values as parameters and return the result. Your code will be much easier to understand and debug.

## 📐 Block 4 — Code Organization and Readability

Writing code that works is the minimum. Writing code that **others (or future you) can understand** is the actual goal.

### Docstrings — documenting your functions

````python
def calculate_average(grades, round_to=2):
    """
    Calculates the average of a list of grades.

    Args:
        grades   (list) : List of numeric grade values.
        round_to (int)  : Number of decimal places to round to. Default is 2.

    Returns:
        float: The rounded average.

    Example:
        >>> calculate_average([3.5, 2.8, 4.0])
        3.43
    """
    return round(sum(grades) / len(grades), round_to)

# Python can show you the docstring on demand
help(calculate_average)
````

### Naming conventions

Good names make code self-documenting. You should rarely need a comment to explain *what* the code does — only *why*.

| What | Convention | Good example | Bad example |
|------|-----------|--------------|-------------|
| Variables | `snake_case` | `monthly_income` | `mi`, `x1`, `temp2` |
| Functions | `snake_case` | `calculate_average` | `calc`, `doStuff` |
| Constants | `UPPER_CASE` | `PASSING_THRESHOLD` | `threshold`, `t` |
| Booleans | Start with `is_`, `has_`, `can_` | `is_enrolled`, `has_debt` | `enrolled`, `flag` |

````python
# Bad — hard to understand without context
def proc(d, t=2.0):
    r = []
    for x in d:
        if x["g"] >= t:
            r.append(x)
    return r


# Good — reads almost like English
PASSING_THRESHOLD = 2.0

def filter_passing_students(students, threshold=PASSING_THRESHOLD):
    """Returns only students whose GPA meets or exceeds the threshold."""
    passing_students = []
    for student in students:
        if student["gpa"] >= threshold:
            passing_students.append(student)
    return passing_students


students = [
    {"name": "Alice", "gpa": 3.8},
    {"name": "Bob",   "gpa": 1.5},
    {"name": "Carol", "gpa": 2.7},
]

result = filter_passing_students(students)
for s in result:
    print(s["name"])
````

### One function, one job

A function should do **exactly one thing**. If you find yourself writing `and` in a function name, split it into two functions.

````python
# Bad — does too many things at once
def process_and_print_and_filter(students):
    total = 0
    for s in students:
        total += s["gpa"]
        if s["gpa"] >= 2.0:
            print(s["name"])   # Mixing calculation with output
    print(f"Average: {total / len(students):.2f}")


# Good — each function has a single, clear responsibility
def calculate_class_average(students):
    """Returns the average GPA of all students."""
    return sum(s["gpa"] for s in students) / len(students)

def filter_passing_students(students, threshold=2.0):
    """Returns students whose GPA meets the threshold."""
    return [s for s in students if s["gpa"] >= threshold]

def print_student_names(students):
    """Prints the name of each student in the list."""
    for s in students:
        print(s["name"])


students = [
    {"name": "Alice", "gpa": 3.8},
    {"name": "Bob",   "gpa": 1.5},
    {"name": "Carol", "gpa": 2.7},
]

print(f"Class average: {calculate_class_average(students):.2f}")
print("Passing students:")
print_student_names(filter_passing_students(students))
````

### Why test your code?

You just wrote five functions. How do you know they work correctly?  
Right now you are checking manually — running them and eyeballing the output. That breaks the moment:
- You change one function and accidentally break another
- Your dataset changes and an edge case appears
- Someone else runs your code on their machine

**Tests are automated checks.** You write them once and run them every time your code changes. If something breaks, you know immediately — and exactly where.

### assert — the building block of every test

`assert` checks that something is `True`. If it is not, Python raises an `AssertionError`.

````python
# assert passes silently when True
assert 2 + 2 == 4
assert "hello".upper() == "HELLO"
print("All assertions passed.")

# assert raises an error when False — read the message
assert 2 + 2 == 5, "Math is broken"
````

````python
# Using assert to verify our functions from this module
def calculate_average(values):
    """Returns the average of a list of numbers. Returns 0 if empty."""
    if not values:
        return 0
    return sum(values) / len(values)

# Test: normal case
assert calculate_average([4.0, 2.0]) == 3.0

# Test: single value
assert calculate_average([3.5]) == 3.5

# Test: empty list
assert calculate_average([]) == 0

# Test: all zeros
assert calculate_average([0, 0, 0]) == 0

print("All tests passed ✅")
````

## 📋 Module 4 Summary

| Concept | Syntax | Key point |
|---------|--------|-----------|
| Define function | `def name(params):` | Code runs only when called |
| Return value | `return result` | Without it the function returns `None` |
| Default parameter | `def f(x, n=2):` | Caller can omit it |
| Keyword argument | `f(x=3, n=4)` | Order does not matter |
| Multiple returns | `return a, b, c` | Unpacked as `a, b, c = f()` |
| Local scope | Variable inside function | Invisible outside |
| Global scope | Variable outside function | Readable inside, avoid modifying |
| Docstring | `""" ... """` | First line inside the function |
| One job rule | — | If you write `and` in the name, split it |
| `assert` | `assert x == y` | Verifies a condition is True |
| pytest test | `def test_..():` in `test_*.py` | Automated, repeatable checks |
