## 🔷 Block 3 — Selection and Filtering

### Analogy: AutoFilter in Excel

When you use Excel's AutoFilter you click a column header and choose a condition.
Pandas does the same thing — but you write the condition in one Python expression.
No clicking, no menus, and it works on millions of rows just as fast.

### Bridge from Module 4

In Module 4 you wrote this to filter employees:
```python
def filter_by(records, key, value):
    result = []
    for record in records:
        if record[key] == value:
            result.append(record)
    return result

filter_by(employees_list, 'department', 'Engineering')
```

Pandas replaces that entire function with one expression:
```python
employees[employees['department'] == 'Engineering']
```

You built `filter_by()` by hand so you understand what Pandas is doing underneath.

### Selecting columns

````python
# Single column → returns a Series.
employees["salary"]
````

````python
# Multiple columns → pass a list of names → returns a DataFrame.
employees[["name", "salary", "department"]]
````

### Boolean filtering — selecting rows by condition

The pattern always has two steps:
1. Create a **boolean Series** (True/False for every row)
2. Pass it inside `df[...]` to keep only the `True` rows

````python
# Step 1: the boolean Series — True where salary >= 4 000 000.
mask = employees["salary"] >= 4000000
print('Boolean mask:')
print(mask.values)
print(f'Rows where True: {mask.sum()}')
````

````python
# Step 2: use the mask to select rows.
high_earners = employees[employees["salary"] >= 4000000]
high_earners
````

### Compound filters — combining conditions

Use `&` (and) and `|` (or) between conditions.
Each condition **must** be wrapped in parentheses.

````python
# Engineers earning at least 4 500 000 COP.
senior_engineers = employees[
    (employees["department"] == "Engineering") &
    (employees["salary"] >= 4500000)
]
senior_engineers
````

### Deliberate error — using `and` instead of `&`

> **Before running:** Python has both `and` and `&`. Which one works here?

````python
# Using Python's 'and' with Pandas Series raises a ValueError.
# Uncomment the lines below to see the error.
#
# broken = employees[
#     employees["department"] == "Engineering"
#     and
#     employees["salary"] >= 4500000
# ]
#
# ValueError: The truth value of a Series is ambiguous.
#
# Fix: use & and wrap each condition in parentheses.
correct = employees[
    (employees["department"] == "Engineering") &
    (employees["salary"] >= 4500000)
]
print(f"Rows found: {len(correct)}")
````

### loc and iloc — selecting by label and by position

| Method | Selects by | Example |
|--------|-----------|---------|
| `.loc[row, col]` | **Label** (row index label, column name) | `df.loc[0, 'name']` |
| `.iloc[row, col]` | **Position** (integer 0, 1, 2 …) | `df.iloc[0, 0]` |

When the index is the default 0, 1, 2 … both give the same result for rows.
Use `.loc` when you have meaningful row labels (e.g. dates, names).
Use `.iloc` when you think in positions (first row, last column, etc.).

````python
# .loc — label-based.
# Row 2, column 'name'.
print(employees.loc[2, "name"])

# Rows 0-2, columns name and salary.
employees.loc[0:2, ["name", "salary"]]
````

````python
# .iloc — position-based.
# First row, first column (position 0, 0).
print(employees.iloc[0, 0])

# First 3 rows, first 2 columns.
employees.iloc[0:3, 0:2]
````

### isin() and between() — convenient shortcuts

````python
# isin() — keep rows where city is Bogotá OR Cali.
bogota_cali = employees[employees["city"].isin(["Bogotá", "Cali"])]
bogota_cali
````

````python
# between() — keep rows where salary is between 3 000 000 and 4 000 000 (inclusive).
mid_range = employees[employees["salary"].between(3000000, 4000000)]
mid_range
````

### apply() — running a function on every row

`.apply()` takes a function and calls it once for each value in a column.
This connects directly to the functions you wrote in Module 4.

````python
# New column: classify each employee as Junior or Senior based on experience.
# apply() runs the function once per row.
def fn(x):
    # if x >= 5:
    #   return "Senior"
    # else:
    #   return "Junior"
    return "Senior" if x >= 5 else "Junior"
    

employees["level"] = employees["years_exp"].apply(fn)

employees[["name", "years_exp", "level"]].head(8)
````

### Connecting apply() to your Module 4 functions

This is the direct bridge between M4 and M6.

````python
# A named function — exactly like Module 4.
def salary_band(salary):
    """Classifies a salary into Low, Mid, or High band."""
    if salary < 3500000:
        return "Low"
    elif salary < 5000000:
        return "Mid"
    else:
        return "High"

# Apply it to the salary column.
employees["band"] = employees["salary"].apply(salary_band)

employees[["name", "salary", "band"]].head(8)
````

### ✏️ Exercise 3 — Add derived columns

**🧠 Before you code — apply the 5 steps:**

- **Input:** the employees DataFrame (with the columns added in Block 4 so far)
- **Output:** the same DataFrame with two new columns: `monthly_bonus` and
  `total_compensation`; sorted by `total_compensation` descending

**Decompose:**
1. For each employee compute a monthly bonus:
   Engineering employees get 10% of salary, all others get 8%
2. For each employee compute total compensation = salary + monthly bonus
3. Sort the DataFrame by total compensation from highest to lowest
4. Display the name, department, salary, bonus, and total compensation columns

**Pseudocode:**
```
For each row: if department is Engineering, bonus is 10 percent of salary,
  otherwise bonus is 8 percent of salary. Store this as a new column.
For each row: add salary and bonus together. Store this as a new column.
Sort the entire table so the highest total compensation appears first.
Display only the relevant columns.
```

**Pattern:** Pandas one-liner (column creation with apply + sort)

````python
# TODO: Add monthly_bonus column.
# Engineering → 10% of salary. All others → 8% of salary.
# Hint: use .apply() with a function that checks the department.

def compute_bonus(row):
    """Returns the monthly bonus for a given row."""
    if row["department"]=="Engineering":
        return row["salary"]*0.1
    else:
        return row["salary"]*0.08

# TODO: apply compute_bonus across rows (use axis=1 to pass the full row).
employees["monthly_bonus"] =employees.apply(compute_bonus, axis=1)

# TODO: Add total_compensation column = salary + monthly_bonus.
employees["total_compensation"] =employees["salary"] + employees["monthly_bonus"]

# TODO: Sort by total_compensation descending.
result = employees.sort_values("total_compensation", ascending=False)

# TODO: Display name, department, salary, monthly_bonus, total_compensation.
result[["name","department","salary","monthly_bonus","total_compensation"]]
````

## 🔷 Block 5 — Mini-Project: Employee Analysis Dashboard

You will use everything from Blocks 1–4 to build a small analysis report.
Each part has a plain-English plan — translate it to Pandas one step at a time.

### Full decomposition — read this before writing any code

**PART A — Filter and explore**
```
Select only the Engineering employees from the table.
Count how many Engineering employees there are.
Calculate the average salary of Engineering employees.
```

**PART B — Add derived columns**
```
Add a seniority column to the FULL employees table:
  fewer than 3 years → Junior
  3 to 5 years (inclusive) → Mid
  more than 5 years → Senior
Add an annual salary column = monthly salary multiplied by 12.
```

**PART C — Summary by department**
```
For each department, compute the average salary.
For each department, count the number of employees.
This is exactly what group_by() + calculate_average() from Module 4 did —
  but Pandas does it in one line with groupby().
```

**PART D — Print the report**
```
Show the top 3 highest-paid employees (name, department, salary).
Show which department has the highest average salary.
```

### Part A — Filter and explore Engineering employees

````python
# TODO Part A: Filter Engineering employees.
# Then compute and print: count and average salary.

# Step 1: filter.
engineers = employees[employees["department"] == "Engineering"]

# Step 2: count.
count = len(engineers)

# Step 3: average salary.
avg_salary = engineers["salary"].mean()

print(f'Engineering headcount  : {count}')
print(f'Average salary (COP)   : {avg_salary:,.0f}')
````

### Part B — Add derived columns

````python
# TODO Part B: Add seniority and annual_salary columns.

def classify_seniority(years):
    """Returns Junior, Mid, or Senior based on years of experience."""
    if years < 3:
        return "Junior"
    elif years <= 5:
        return "Mid"
    else:
        return "Senior"

# Apply to years_exp column.
employees["seniority"] = employees["years_exp"].apply(classify_seniority)

# Annual salary = monthly salary × 12.
employees["annual_salary"] = employees["salary"]*12

# Preview.
employees[["name", "years_exp", "seniority", "salary", "annual_salary"]].head(8)
````

### Part C — Summary by department

````python
# TODO Part C: Group by department and compute average salary and headcount.

# Average salary per department.
avg_by_dept = employees.groupby("department")["salary"].mean()

# Headcount per department.
count_by_dept = employees.groupby("department")["salary"].count()

print('Average salary by department:')
print(avg_by_dept)
print()
print('Headcount by department:')
print(count_by_dept)
````

### Part D — Print the report

````python
# TODO Part D: Top 3 highest-paid employees and highest-paying department.

# Top 3 highest-paid employees.
top3 = employees.sort_values("salary", ascending=False).head(3)

print('Top 3 highest-paid employees:')
# TODO: print each of the top 3 with name, department, salary.
print(top3[["name", "department", "salary"]])

# Department with highest average salary.
# Hint: use avg_by_dept from Part C and .idxmax().
best_dept = avg_by_dept.idxmax()
print()
print(f'Highest average salary department: {best_dept}')
````

## 📋 Module 6 Summary

| Concept | Syntax | Key point |
|---------|--------|-----------|
| Import Pandas | `import pandas as pd` | Standard alias is `pd` |
| Create DataFrame | `pd.DataFrame({...})` | From dict of lists |
| Shape | `df.shape` | (rows, columns) |
| Column types | `df.dtypes` | Each column has one dtype |
| Access column | `df["col"]` | Returns a Series |
| Access multiple cols | `df[["c1", "c2"]]` | Returns a DataFrame |
| First/last rows | `df.head(n)` / `df.tail(n)` | Default n = 5 |
| Overview | `df.info()` | Dtypes + non-null counts |
| Statistics | `df.describe()` | Numeric columns only |
| Frequency table | `df["col"].value_counts()` | Most frequent first |
| Boolean filter | `df[df["col"] > x]` | Returns matching rows |
| Compound filter | `(cond1) & (cond2)` | Use `&` not `and` |
| Filter shortcuts | `.isin([...])`, `.between(a, b)` | Cleaner than long conditions |
| Label selection | `df.loc[row, col]` | By row label + col name |
| Position selection | `df.iloc[row, col]` | By integer position |
| New column | `df["new"] = expr` | Applied to every row |
| Apply function | `df["col"].apply(fn)` | fn called once per value |
| Rename columns | `df.rename(columns={"old": "new"})` | Returns new DataFrame |
| Sort | `df.sort_values("col", ascending=False)` | Returns new DataFrame |
| Group and aggregate | `df.groupby("col")["val"].mean()` | Replaces M4's `group_by()` |
