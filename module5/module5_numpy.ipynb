## 🔷 Block 1 — From Lists to Arrays

### The problem with plain Python lists

You already know how to store numbers in a list. But what happens when you need to apply the same operation to every element?

````python
# Apply 19% IVA tax to a list of prices — the plain Python way.
prices = [2500000, 850000, 1200000, 420000, 350000]

prices_with_tax = []
for p in prices:
    prices_with_tax.append(p * 1.19)
# Round 1: p = 2500000  →  2500000 * 1.19 = 2975000.0
# Round 2: p = 850000   →  850000  * 1.19 = 1011500.0
# Round 3: p = 1200000  →  1200000 * 1.19 = 1428000.0
# Round 4: p = 420000   →  420000  * 1.19 = 499800.0
# Round 5: p = 350000   →  350000  * 1.19 = 416500.0

print(prices_with_tax)
````

### Analogy: backpack vs. train car

A **Python list** is like a backpack — it holds anything: numbers, strings, booleans, even other lists. Very flexible, but operations happen one item at a time.

A **NumPy array** is like a train car — every seat holds the same type of passenger (all integers, or all floats). Because all seats are identical, the train can move all passengers at once — operations run in **parallel**, with no Python loop needed.

This is called **vectorization**.

````python
import numpy as np

# The NumPy way — one line, no loop.
prices = np.array([2500000, 850000, 1200000, 420000, 350000])
prices_with_tax = prices * 1.19

print(prices_with_tax)

# Same result, but NumPy did the multiplication in compiled C code — much faster
# on large datasets (millions of rows).
````

## 🔷 Block 2 — Vectorized Operations

### Arithmetic on arrays

When you apply an arithmetic operator to a NumPy array, it applies to **every element** — no loop needed. This is called **element-wise** operation.

````python
import numpy as np

# Two arrays of the same size.
units_sold  = np.array([45,  120, 30,  200, 15])
unit_prices = np.array([2500000, 1200000, 800000, 350000, 900000])

# Element-wise multiplication — position 0 * position 0, etc.
revenue = units_sold * unit_prices
print(f"Revenue per product: {revenue}")

# You can also combine with scalar arithmetic.
revenue_with_tax = revenue * 1.19
print(f"Revenue with IVA:    {revenue_with_tax}")
````

### Analogy: broadcasting

**Broadcasting** is like a chef seasoning an entire tray of food at once instead of picking up each piece individually and adding salt.

When you write `prices * 1.19`, NumPy "broadcasts" the scalar `1.19` across every element — the scalar is applied to all items in one step.

````python
import numpy as np

salaries = np.array([3200000, 4500000, 5100000, 2900000, 4800000])

# Add a 10% bonus to all salaries at once.
salaries_with_bonus = salaries * 1.10
print(f"Original salaries : {salaries}")
print(f"After 10% bonus   : {salaries_with_bonus}")

# Subtract social security (4%) from all at once.
net_salaries = salaries_with_bonus * (1 - 0.04)
print(f"Net salaries (4% deducted): {net_salaries}")
````

### Comparison operators → boolean arrays

You can compare an array against a value. NumPy returns a **boolean array** — one `True`/`False` per element.

````python
import numpy as np

grades = np.array([4.5, 2.1, 3.7, 1.8, 4.8, 3.0, 2.9])

# Which grades are passing (>= 3.0)?
is_passing = grades >= 3.0
print(f"Grades        : {grades}")
print(f"Passing mask  : {is_passing}")
# True where the condition is met, False where it is not.
````

### Boolean indexing — replacing the Filter pattern

In Module 3 you learned the Filter pattern:
```
result = []  →  loop  →  if condition: result.append(x)
```

NumPy collapses that entire pattern into **one line**: pass a boolean array as the index.

> This does not mean you will never use the Filter pattern — but for numerical data, boolean indexing is faster and more readable.

````python
import numpy as np

grades = np.array([4.5, 2.1, 3.7, 1.8, 4.8, 3.0, 2.9])

# --- Old way (Filter pattern with a loop) ---
passing_loop = []
for g in grades:
    if g >= 3.0:
        passing_loop.append(g)
# Round 1: g = 4.5  →  4.5 >= 3.0 is True   →  passing_loop = [4.5]
# Round 2: g = 2.1  →  2.1 >= 3.0 is False  →  passing_loop = [4.5]
# Round 3: g = 3.7  →  3.7 >= 3.0 is True   →  passing_loop = [4.5, 3.7]
# Round 4: g = 1.8  →  1.8 >= 3.0 is False  →  passing_loop = [4.5, 3.7]
# Round 5: g = 4.8  →  4.8 >= 3.0 is True   →  passing_loop = [4.5, 3.7, 4.8]
# Round 6: g = 3.0  →  3.0 >= 3.0 is True   →  passing_loop = [4.5, 3.7, 4.8, 3.0]
# Round 7: g = 2.9  →  2.9 >= 3.0 is False  →  passing_loop = [4.5, 3.7, 4.8, 3.0]
print(f"Loop result    : {passing_loop}")

# --- NumPy way (boolean indexing — one line) ---
passing_numpy = grades[grades >= 3.0]
print(f"NumPy result   : {passing_numpy}")
````

### 🚨 Deliberate error — mismatched array sizes

Read the error before looking at the fix.

````python
import numpy as np

# Two arrays of DIFFERENT sizes.
prices = np.array([2500000, 850000, 1200000, 420000])
quantities = np.array([10, 5, 20])   # Only 3 elements — wrong!

# NumPy cannot add element-wise when sizes differ.
total = prices + quantities
````

**What went wrong?** NumPy requires both arrays to have the same shape for element-wise operations. There is no "auto-fill" for missing values.

**Fix:** make sure both arrays have the same length.

````python
import numpy as np

prices     = np.array([2500000, 850000, 1200000, 420000])
quantities = np.array([10, 5, 20, 8])   # Now 4 elements — matches.

total = prices * quantities
print(f"Total value per product: {total}")
````

## 🔷 Block 3 — Indexing and Slicing

### 1D indexing — same rules as lists

If you already know how to index a list, you already know how to index a 1D NumPy array.

````python
import numpy as np

grades = np.array([4.5, 2.1, 3.7, 1.8, 4.8, 3.0, 2.9])

print(f"First grade     : {grades[0]}")      # Index 0.
print(f"Last grade      : {grades[-1]}")     # Index -1 = last.
print(f"Grades 1 to 3   : {grades[1:4]}")    # Slice — stop excluded.
print(f"Every other one : {grades[::2]}")    # Step = 2.
print(f"Reversed        : {grades[::-1]}")   # Step = -1.
````

### 2D arrays — the spreadsheet model

A 2D array has **rows** and **columns**. Think of it as a spreadsheet:
- Rows are observations (e.g., products)
- Columns are variables (e.g., months)

You index with `arr[row, col]` — **row first, then column**.

```
          Month 0  Month 1  Month 2  Month 3
Product A   120      135       98      150
Product B   200      180      220      195
Product C    45       60       55       70
```

````python
import numpy as np

# 3 products × 4 months of sales (units sold).
sales = np.array([
    [120, 135,  98, 150],   # Product A
    [200, 180, 220, 195],   # Product B
    [ 45,  60,  55,  70],   # Product C
])

print(f"Shape  : {sales.shape}")   # (3 rows, 4 columns)
print(f"ndim   : {sales.ndim}")    # 2 dimensions.
print(f"size   : {sales.size}")    # 3 * 4 = 12 total elements.
````

````python
import numpy as np

sales = np.array([
    [120, 135,  98, 150],   # Row 0 — Product A
    [200, 180, 220, 195],   # Row 1 — Product B
    [ 45,  60,  55,  70],   # Row 2 — Product C
])

# Single element — row 1, column 2 (Product B in month 2).
print(f"Product B, Month 2      : {sales[1, 2]}")

# Entire row 0 — all months for Product A.
print(f"Product A (all months)  : {sales[0, :]}")

# Entire column 2 — all products in month 2.
print(f"All products in month 2 : {sales[:, 2]}")

# Submatrix — rows 0 and 1, months 0 and 1.
print(f"Top-left 2x2 block:\n{sales[0:2, 0:2]}")
````

### Fancy indexing — selecting non-consecutive rows or columns

You can pass a **list of indices** instead of a slice to pick specific rows or columns in any order.

````python
import numpy as np

sales = np.array([
    [120, 135,  98, 150],   # Row 0 — Product A
    [200, 180, 220, 195],   # Row 1 — Product B
    [ 45,  60,  55,  70],   # Row 2 — Product C
])

# Select rows 0 and 2 (Product A and Product C) — skip Product B.
selection = sales[[0, 2], :]
print(f"Products A and C:\n{selection}")

# Select columns 1 and 3 (months 1 and 3 only).
months_1_3 = sales[:, [1, 3]]
print(f"Months 1 and 3:\n{months_1_3}")
````

## 🔷 Block 4 — Statistical Functions

NumPy provides fast built-in functions for descriptive statistics. You will use these constantly in data analysis.

````python
import numpy as np

# Salaries for 6 employees (COP).
salaries = np.array([3200000, 4500000, 5100000, 2900000, 4800000, 3750000])

print(f"Sum        : {np.sum(salaries):,.0f} COP")
print(f"Mean       : {np.mean(salaries):,.0f} COP")
print(f"Median     : {np.median(salaries):,.0f} COP")
print(f"Std dev    : {np.std(salaries):,.0f} COP")
print(f"Min        : {np.min(salaries):,.0f} COP")
print(f"Max        : {np.max(salaries):,.0f} COP")
print(f"75th pct   : {np.percentile(salaries, 75):,.0f} COP")
````

### argmax and argmin — finding the position of extreme values

`np.argmax()` and `np.argmin()` return the **index** of the maximum or minimum value — not the value itself. This is useful for looking up corresponding labels.

````python
import numpy as np

employees = ["Valentina", "Santiago", "Camila", "Andrés", "Daniela", "Felipe"]
salaries  = np.array([3200000, 4500000, 5100000, 2900000, 4800000, 3750000])

highest_idx = np.argmax(salaries)
lowest_idx  = np.argmin(salaries)

print(f"Highest paid : {employees[highest_idx]} ({salaries[highest_idx]:,} COP)")
print(f"Lowest paid  : {employees[lowest_idx]}  ({salaries[lowest_idx]:,} COP)")
````

### The axis parameter — operating on rows or columns of a 2D array

When you call `np.sum(arr)` on a 2D array without `axis`, it sums **all elements** into one number. The `axis` parameter lets you choose a direction:

- `axis=0` — **collapse rows** (operate vertically, result has one value per column)
- `axis=1` — **collapse columns** (operate horizontally, result has one value per row)

**Analogy:** imagine pressing the array flat like a pancake.
- `axis=0`: press it **downward** — each column gets squashed into one number.
- `axis=1`: press it **sideways** — each row gets squashed into one number.

````python
import numpy as np

# 3 products × 4 months of sales.
sales = np.array([
    [120, 135,  98, 150],   # Product A
    [200, 180, 220, 195],   # Product B
    [ 45,  60,  55,  70],   # Product C
])

# No axis — single total for the entire array.
total_all = np.sum(sales)
print(f"Grand total (all products, all months) : {total_all}")

# axis=1 → sum across columns → one total per ROW (per product).
total_per_product = np.sum(sales, axis=1)
print(f"Total per product (axis=1)             : {total_per_product}")
# Product A: 120+135+98+150 = 503
# Product B: 200+180+220+195 = 795
# Product C: 45+60+55+70 = 230

# axis=0 → sum across rows → one total per COLUMN (per month).
total_per_month = np.sum(sales, axis=0)
print(f"Total per month   (axis=0)             : {total_per_month}")
# Month 0: 120+200+45 = 365
# Month 1: 135+180+60 = 375
# Month 2: 98+220+55  = 373
# Month 3: 150+195+70 = 415
````

````python
import numpy as np

sales = np.array([
    [120, 135,  98, 150],
    [200, 180, 220, 195],
    [ 45,  60,  55,  70],
])

# Average sales per product (across all 4 months).
avg_per_product = np.mean(sales, axis=1)
print(f"Avg per product : {avg_per_product}")

# Average sales per month (across all 3 products).
avg_per_month = np.mean(sales, axis=0)
print(f"Avg per month   : {avg_per_month}")

# Which product has the highest average?
best_product_idx = np.argmax(avg_per_product)
print(f"Best product index: {best_product_idx}  (avg = {avg_per_product[best_product_idx]:.1f} units)")
````

## 📋 Module 5 Summary

| Concept | Syntax | Key point |
|---------|--------|-----------|
| Import NumPy | `import numpy as np` | Always aliased as `np` |
| Create array | `np.array([1, 2, 3])` | All elements must be the same type |
| Array attributes | `.dtype`, `.shape`, `.ndim`, `.size` | Inspect before operating |
| Zeros / ones | `np.zeros(n)`, `np.ones(n)` | Useful as initial containers |
| Range array | `np.arange(start, stop, step)` | Stop is excluded |
| Evenly spaced | `np.linspace(start, stop, n)` | Stop is included; n = count of points |
| Vectorized math | `arr * 1.19` | No loop needed; applied to every element |
| Broadcasting | `scalar op array` | Scalar is applied to all elements |
| Boolean array | `arr >= 3.0` | Returns True/False per element |
| Boolean indexing | `arr[arr >= 3.0]` | Replaces the Filter pattern for numbers |
| 2D indexing | `arr[row, col]` | Row first, then column |
| Row slice | `arr[i, :]` | All columns of row i |
| Column slice | `arr[:, j]` | All rows of column j |
| Fancy indexing | `arr[[0, 2], :]` | Non-consecutive rows or columns |
| Sum / mean | `np.sum(arr)`, `np.mean(arr)` | Works on full array or per axis |
| Std / median | `np.std(arr)`, `np.median(arr)` | Descriptive statistics |
| Percentile | `np.percentile(arr, 75)` | 75th percentile |
| Position of max/min | `np.argmax(arr)`, `np.argmin(arr)` | Returns index, not value |
| Axis parameter | `np.sum(arr, axis=0)` | 0 = collapse rows; 1 = collapse columns |
